    Clinical Context
</div>
Pneumonia is an infection that inflames the air sacs in one or both lungs.
On a chest X-ray, pneumonia appears as white cloudy areas (opacities)
in the lung fields — caused by fluid or pus filling the air sacs.<br><br>
A normal chest X-ray shows clear, dark lung fields with no opacities.
The visual difference between normal and pneumonia X-rays
is subtle enough that even trained radiologists can disagree —
making this an ideal task for a CNN that can detect
spatial patterns invisible to the naked eye.
</div>

<h3 style="color: #8B1E3F; margin: 0 0 14px 0; font-size: 18px;">
    Why Not a Regular Neural Network?
</h3>

A standard fully connected network (MLP) treats every input
as a flat vector of numbers — no concept of spatial structure.<br><br>

A 224×224 grayscale X-ray image contains
<strong>50,176 pixels</strong>. Flattening it into a vector
and connecting every pixel to every neuron in the first layer
would require millions of parameters — just for one layer.<br><br>

More critically, an MLP has <strong>no concept of locality</strong> —
it does not know that two neighboring pixels are more related
than two pixels at opposite corners of the image.
A small white opacity in the lower right lung field
carries the same diagnostic weight wherever it appears —
but an MLP has no way to learn this translation invariance.<br><br>

<h3 style="color: #8B1E3F; margin: 0 0 14px 0; font-size: 18px;">

Raw pixel values range from 0 to 255.

the right scale for weights when inputs vary widely.<br><br>

The solution is <strong>normalization</strong> — scaling pixel values
to a small, consistent range:

**Count Images**

```python

# Section 4 — Dataset Audit

#  Count images per split per class
splits = {'train': TRAIN_DIR, 'val': VAL_DIR, 'test': TEST_DIR}
classes = ['NORMAL', 'PNEUMONIA']

print("Image counts per split:")
print("=" * 45)
print(f"{'Split':<10} {'NORMAL':>10} {'PNEUMONIA':>12} {'Total':>8}")
print("-" * 45)

counts = {}
for split, path in splits.items():
    split_counts = {}
    for cls in classes:
        cls_path = os.path.join(path, cls)
        n = len([f for f in os.listdir(cls_path)
                 if f.lower().endswith(('.jpg', '.jpeg', '.png'))])
        split_counts[cls] = n
    counts[split] = split_counts
    total = sum(split_counts.values())
    print(f"{split:<10} {split_counts['NORMAL']:>10} "
          f"{split_counts['PNEUMONIA']:>12} {total:>8}")

print("-" * 45)
total_all = sum(sum(v.values()) for v in counts.values())
print(f"{'TOTAL':<10} "
      f"{sum(counts[s]['NORMAL'] for s in counts):>10} "
      f"{sum(counts[s]['PNEUMONIA'] for s in counts):>12} "
      f"{total_all:>8}")
```

<h3 style="color: #8B1E3F; margin: 0 0 14px 0; font-size: 18px;">

OpenCV reads images in <strong>BGR order</strong>
(Blue, Green, Red) — the opposite of the standard RGB used
by Matplotlib and TensorFlow.<br><br>

If an image is displayed without converting from BGR to RGB,
the red and blue channels are swapped —
colors look wrong. For grayscale X-rays this does not
cause a visible problem (all channels are identical),
but the conversion is good practice and essential for color images.<br><br>

**Define Preprocessing Configuration**

```python
# Section 6 — Image Preprocessing Pipeline


#  Preprocessing configuration
IMG_SIZE = (150, 150)
PIXEL_SCALE = 255.0

print("Preprocessing configuration")
print("=" * 45)
print(f"Target image size : {IMG_SIZE}")
print(f"Normalization     : [0, 255] → [0, 1]")
print(f"Pixel scale       : {PIXEL_SCALE}")
```

**Build Preprocessing Function**

```python

# Complete OpenCV preprocessing function
def preprocess_image(image_path, target_size=IMG_SIZE):
    """
    Load and preprocess a chest X-ray image.

    Steps:
        1. Read image with OpenCV.
        2. Convert BGR to RGB.
        3. Resize to target dimensions.
        4. Convert to float32.
        5. Normalize pixel values to [0, 1].

    Returns:
        np.ndarray: Preprocessed image.
    """

    # 1. Read image
    image = cv2.imread(image_path)

    if image is None:
        raise ValueError(f"Could not read image: {image_path}")

    # 2. Convert BGR → RGB
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # 3. Resize
    image = cv2.resize(image, target_size)

    # 4. Convert to float32
    image = image.astype(np.float32)

    # 5. Normalize to [0, 1]
    image = image / PIXEL_SCALE

    return image
```

Deep learning models can easily memorize the training images when the

This can lead to <strong style="color: #C44569;">overfitting</strong>:

<p>

<strong style="color: #C44569;">Data augmentation</strong> addresses this by
applying controlled transformations to training images, creating additional
variation without collecting new X-rays.
</p>

<p>
For chest X-rays, however, augmentation must be used carefully. A
transformation should preserve the medically meaningful structures of the
image while introducing realistic variation. Aggressive or inappropriate
transformations could change the visual characteristics that the model needs
to learn.

<h3 style="color: #8B1E3F; margin-top: 0;">
    Why use data generators?
</h3>

<p>
A CNN should not receive the entire image dataset as one large object
during training. Instead, images are loaded and processed in
<strong style="color:#C44569;">small batches</strong>.
A data generator handles this process automatically.
</p>

<p>
For each batch, the generator loads images from disk, resizes them to
the required dimensions, applies the appropriate preprocessing, and
returns the images together with their labels.
</p>

<p>
For the training pipeline, augmentation is also applied dynamically.
This means the same original image can appear with slightly different
transformations across different training iterations.
</p>

**Define CNN Architecture**

```python
cnn_model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(*IMG_SIZE, 3)),

    # Feature extraction
    tf.keras.layers.Conv2D(32, (3, 3), activation="relu"),
    tf.keras.layers.MaxPooling2D((2, 2)),

    tf.keras.layers.Conv2D(64, (3, 3), activation="relu"),
    tf.keras.layers.MaxPooling2D((2, 2)),

    tf.keras.layers.Conv2D(128, (3, 3), activation="relu"),
    tf.keras.layers.MaxPooling2D((2, 2)),

    # Classification
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dropout(0.5),
    tf.keras.layers.Dense(1, activation="sigmoid")
])

print("CNN model created successfully")
```

**Model Summary**

```python
print("CNN Model Architecture")
print("=" * 55)

cnn_model.summary()
```

**Compile CNN**

```python
cnn_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.AUC(name="auc")
    ]
)

print("✓ CNN model compiled successfully")
print()
print("Optimizer : Adam")
print("Learning rate : 0.001")
print("Loss : Binary Crossentropy")
print("Metrics : Accuracy, AUC")
```

the model has a defined input shape, feature extraction layers,

**Training callbacks**

```python
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=2,
    min_lr=1e-6
)

print("✓ Training callbacks configured")
print()
print("EarlyStopping:")
print("  Monitor           : val_loss")
print("  Patience          : 5")
print("  Restore best      : True")

print()
print("ReduceLROnPlateau:")
print("  Monitor           : val_loss")
print("  Factor            : 0.5")
print("  Patience          : 2")
print("  Minimum LR        : 1e-6")
```

**Train CNN**

```python
EPOCHS = 30

history = cnn_model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=EPOCHS,
    callbacks=[early_stopping, reduce_lr],
    verbose=1
)
```

<strong style="color: #8B1E3F;">Why Transfer Learning?</strong><br><br>

Instead of learning visual features from scratch, transfer learning
starts with a model that has already learned general image features
from a large dataset. These learned representations can then be
adapted to a new classification task with much less training data
and computational cost.<br><br>

<strong style="color: #C44569;">MobileNetV2</strong> is a lightweight
CNN architecture designed to provide efficient image feature
extraction while keeping the number of parameters relatively small.
For this experiment, the pretrained convolutional base will be used
as a feature extractor, while a new classification head will be
trained for the <code>NORMAL</code> vs <code>PNEUMONIA</code> task.
</div>

```python
id="s12mobilenet"
# Load MobileNetV2 pretrained on ImageNet

mobilenet_base = tf.keras.applications.MobileNetV2(
    input_shape=(*IMG_SIZE, 3),
    include_top=False,
    weights="imagenet"
)

mobilenet_base.trainable = False

print("MobileNetV2 Base Model")
print("=" * 50)
print(f"Input shape       : {mobilenet_base.input_shape}")
print(f"Output shape      : {mobilenet_base.output_shape}")
print(f"Total layers      : {len(mobilenet_base.layers)}")
print(f"Trainable         : {mobilenet_base.trainable}")
print(f"Parameters        : {mobilenet_base.count_params():,}")
```

**Build MobileNetV2-specific data generators**

```python


MOBILENET_IMG_SIZE = (224, 224)

mobilenet_train_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input,
    rotation_range=AUGMENTATION_CONFIG["rotation_range"],
    zoom_range=AUGMENTATION_CONFIG["zoom_range"],
    width_shift_range=AUGMENTATION_CONFIG["width_shift_range"],
    height_shift_range=AUGMENTATION_CONFIG["height_shift_range"],
    horizontal_flip=AUGMENTATION_CONFIG["horizontal_flip"],
    brightness_range=AUGMENTATION_CONFIG["brightness_range"],
    validation_split=0.2
)

mobilenet_val_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input,
    validation_split=0.2
)

mobilenet_test_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input
)

print("MobileNetV2 Data Generators")
print("=" * 50)
print(f"Image size        : {MOBILENET_IMG_SIZE}")
print(f"Preprocessing     : MobileNetV2 preprocess_input")
print(f"Expected range    : [-1, 1]")
print(f"Training augment. : Enabled")
print(f"Validation augment.: Disabled")
print(f"Test augment.     : Disabled")
```

**Verify MobileNetV2 training, validation, and test pipelines**

```python


mobilenet_train_generator = mobilenet_train_datagen.flow_from_directory(
    TRAIN_DIR,
    target_size=MOBILENET_IMG_SIZE,
    batch_size=32,
    class_mode="binary",
    subset="training",
    shuffle=True,
    seed=SEED
)

mobilenet_val_generator = mobilenet_val_datagen.flow_from_directory(
    TRAIN_DIR,
    target_size=MOBILENET_IMG_SIZE,
    batch_size=32,
    class_mode="binary",
    subset="validation",
    shuffle=False,
    seed=SEED
)

mobilenet_test_generator = mobilenet_test_datagen.flow_from_directory(
    TEST_DIR,
    target_size=MOBILENET_IMG_SIZE,
    batch_size=32,
    class_mode="binary",
    shuffle=False
)

print("\nMobileNetV2 Pipeline Verification")
print("=" * 55)

print(f"Training samples   : {mobilenet_train_generator.samples}")
print(f"Validation samples : {mobilenet_val_generator.samples}")
print(f"Test samples       : {mobilenet_test_generator.samples}")

print(f"\nClass mapping      : {mobilenet_train_generator.class_indices}")
print(f"Training batches   : {len(mobilenet_train_generator)}")
print(f"Validation batches : {len(mobilenet_val_generator)}")
print(f"Test batches       : {len(mobilenet_test_generator)}")
```

**Build MobileNetV2 transfer learning model**

```python


mobilenet_model = tf.keras.Sequential([
    mobilenet_base,

    tf.keras.layers.GlobalAveragePooling2D(),

    tf.keras.layers.Dropout(0.3),

    tf.keras.layers.Dense(
        1,
        activation="sigmoid"
    )
])

print("✓ MobileNetV2 transfer learning model created successfully")
```

**Inspect the MobileNetV2 transfer learning model**

```python


mobilenet_model.summary()

print("\nTrainable Parameters")
print("=" * 50)

trainable_params = np.sum([
    np.prod(variable.shape)
    for variable in mobilenet_model.trainable_variables
])

non_trainable_params = np.sum([
    np.prod(variable.shape)
    for variable in mobilenet_model.non_trainable_variables
])

print(f"Trainable parameters     : {trainable_params:,}")
print(f"Non-trainable parameters : {non_trainable_params:,}")
print(f"Total parameters         : {trainable_params + non_trainable_params:,}")
```

**MobileNetV2 training callbacks**

```python


mobilenet_early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

mobilenet_reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=2,
    min_lr=1e-6
)

MOBILENET_EPOCHS = 30

print("MobileNetV2 Training Configuration")
print("=" * 50)

print(f"Epochs              : {MOBILENET_EPOCHS}")
print("EarlyStopping       : patience=5")
print("Restore best        : True")
print("ReduceLROnPlateau   : factor=0.5, patience=2")
print("Minimum learning rate: 1e-6")
```

**Rebuild MobileNetV2 with the correct input size**

```python


MOBILENET_IMG_SIZE = (224, 224)

mobilenet_base = tf.keras.applications.MobileNetV2(
    input_shape=(*MOBILENET_IMG_SIZE, 3),
    include_top=False,
    weights="imagenet"
)

mobilenet_base.trainable = False

print("MobileNetV2 Base Model Rebuilt")
print("=" * 55)
print(f"Input shape       : {mobilenet_base.input_shape}")
print(f"Output shape      : {mobilenet_base.output_shape}")
print(f"Total layers      : {len(mobilenet_base.layers)}")
print(f"Trainable         : {mobilenet_base.trainable}")
print(f"Parameters        : {mobilenet_base.count_params():,}")
```

**Rebuild the MobileNetV2 transfer learning model**

```python

# using the corrected 224x224 pretrained base

mobilenet_model = tf.keras.Sequential([
    mobilenet_base,

    tf.keras.layers.GlobalAveragePooling2D(),

    tf.keras.layers.Dropout(0.3),

    tf.keras.layers.Dense(
        1,
        activation="sigmoid"
    )
])

print("✓ MobileNetV2 transfer learning model rebuilt successfully")
print()
print(f"Input shape  : {mobilenet_model.input_shape}")
print(f"Output shape : {mobilenet_model.output_shape}")
```

**Compile the rebuilt MobileNetV2 model**

```python


mobilenet_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.AUC(name="auc")
    ]
)

print("✓ Rebuilt MobileNetV2 model compiled successfully")
print()
print("Optimizer    : Adam")
print("Learning rate: 0.001")
print("Loss         : Binary Crossentropy")
print("Metrics      : Accuracy, AUC")
```

**Train MobileNetV2 with frozen pretrained features**

```python
mobilenet_history = mobilenet_model.fit(
    mobilenet_train_generator,
    validation_data=mobilenet_val_generator,
    epochs=MOBILENET_EPOCHS,
    callbacks=[
        mobilenet_early_stopping,
        mobilenet_reduce_lr
    ],
    verbose=1
)
```

**Generate MobileNetV2 predictions on the untouched test set**

```python

mobilenet_test_generator.reset()

mobilenet_y_true = mobilenet_test_generator.classes

mobilenet_y_prob = mobilenet_model.predict(
    mobilenet_test_generator,
    verbose=1
).ravel()

mobilenet_y_pred = (
    mobilenet_y_prob >= 0.5
).astype(int)

print("MobileNetV2 Test Predictions")
print("=" * 50)
print(f"True labels       : {mobilenet_y_true.shape}")
print(f"Predicted probs   : {mobilenet_y_prob.shape}")
print(f"Predicted labels  : {mobilenet_y_pred.shape}")
print(
    f"Probability range : "
    f"{mobilenet_y_prob.min():.4f} → "
    f"{mobilenet_y_prob.max():.4f}"
)
print("Threshold         : 0.5")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

mobilenet_accuracy = accuracy_score(
    mobilenet_y_true,
    mobilenet_y_pred
)

mobilenet_precision = precision_score(
    mobilenet_y_true,
    mobilenet_y_pred
)

mobilenet_recall = recall_score(
    mobilenet_y_true,
    mobilenet_y_pred
)

mobilenet_f1 = f1_score(
    mobilenet_y_true,
    mobilenet_y_pred
)

print("MobileNetV2 Test Metrics")
print("=" * 50)
print(f"Accuracy  : {mobilenet_accuracy:.4f}")
print(f"Precision : {mobilenet_precision:.4f}")
print(f"Recall    : {mobilenet_recall:.4f}")
print(f"F1-score  : {mobilenet_f1:.4f}")

print("\nClassification Report")
print("=" * 50)

print(
    classification_report(
        mobilenet_y_true,
        mobilenet_y_pred,
        target_names=["NORMAL", "PNEUMONIA"],
        digits=4
    )
)

```

```python
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

# Calculate the confusion matrix
cm = confusion_matrix(
    mobilenet_y_true,
    mobilenet_y_pred
)

# Create a custom colormap matching the notebook style
cm_cmap = LinearSegmentedColormap.from_list(
    "burgundy_pink",
    [
        "#fff5f8",
        "#F8A5C2",
        "#C44569",
        "#8B1E3F"
    ]
)

plt.figure(figsize=(7, 6))

plt.imshow(
    cm,
    interpolation="nearest",
    cmap=cm_cmap
)

plt.title(
    "MobileNetV2 Confusion Matrix",
    fontsize=17,
    fontweight="bold",
    color="#8B1E3F",
    pad=18
)

plt.xlabel(
    "Predicted Label",
    fontsize=12,
    fontweight="bold",
    color="#8B1E3F"
)

plt.ylabel(
    "True Label",
    fontsize=12,
    fontweight="bold",
    color="#8B1E3F"
)

class_names = [
    "NORMAL",
    "PNEUMONIA"
]

plt.xticks(
    range(len(class_names)),
    class_names,
    fontsize=11,
    color="#444"
)

plt.yticks(
    range(len(class_names)),
    class_names,
    fontsize=11,
    color="#444"
)

# Display the number of images inside each cell
threshold = cm.max() / 2

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center",
            fontsize=17,
            fontweight="bold",
            color=(
                "white"
                if cm[i, j] > threshold
                else "#8B1E3F"
            )
        )

cbar = plt.colorbar()

cbar.set_label(
    "Number of Images",
    fontsize=11,
    color="#8B1E3F"
)

plt.tight_layout()
plt.show()
```

<strong>Key Learning:</strong>

A reliable computer vision pipeline starts before model training.
Images must first be audited, standardized, and converted into a
consistent numerical representation. OpenCV provided the foundation
for loading, resizing, color conversion, and normalization, while
augmentation introduced controlled variation into the training data.<br><br>

The CNN from scratch demonstrated how convolutional networks can learn
visual features directly from medical images. However, the
<strong>MobileNetV2 transfer-learning approach achieved stronger test-set
performance</strong>, particularly in AUC, showing the value of
pretrained visual representations.<br><br>

One important engineering observation was the gap between validation
and test performance. MobileNetV2 reached approximately
<strong>97.41% validation accuracy</strong> but achieved
<strong>83.49% test accuracy</strong>. This gap should be investigated
rather than ignored, and it prevents us from treating validation
performance alone as evidence of final model quality.<br><br>

The confusion matrices also showed that both models strongly favored
detecting PNEUMONIA, achieving more than <strong>98% recall</strong>,
but produced a considerable number of false positives among NORMAL
images. Therefore, Accuracy alone is not sufficient for evaluating
this imbalanced medical classification problem. Recall, Precision,
F1-score, AUC, and class-level errors must all be considered.
