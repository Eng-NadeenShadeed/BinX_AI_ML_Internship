<h3 style="color: #8B1E3F; margin: 0 0 14px 0; font-size: 18px;">
    Training vs Serving Preprocessing — A Critical Distinction
</h3>

One of the most common causes of production NLP failures is
<strong>training/serving skew</strong> —
the preprocessing applied at training time differs
from the preprocessing applied when the model sees new data.<br><br>

If the training pipeline lowercases and removes punctuation,
but the prediction pipeline does not —
the model receives inputs it has never seen during training
and performance degrades silently.<br><br>

**Preprocessing Pipeline**

```python

# Section 3 — Preprocessing Pipeline
# Identical to Day 1 — single source of truth


lemmatizer   = WordNetLemmatizer()
stop_words   = set(stopwords.words('english'))
negations    = {'no','not','nor','never','neither',
                'nobody','nothing','nowhere','none'}
custom_stops = stop_words - negations

def preprocess(text):
    """
    Full NLP preprocessing pipeline.
    Must be applied identically at training, validation,
    test, and prediction time to avoid training/serving skew.
    """
    text   = str(text).lower()
    text   = text.replace('-', ' ')
    text   = re.sub(r'[^a-z\s]', '', text)
    text   = re.sub(r'\s+', ' ', text).strip()
    tokens = word_tokenize(text)
    tokens = [t for t in tokens if t not in custom_stops]
    tokens = [lemmatizer.lemmatize(t, pos='v') for t in tokens]
    return ' '.join(tokens)

print("Applying preprocessing pipeline...")
df['cleaned'] = df['transcription'].apply(preprocess)
print(f"✓ Done — {len(df)} rows processed")
print()
print("Sample — original vs cleaned:")
print("-" * 50)
print("Original:", df['transcription'].iloc[0][:80])
print("Cleaned :", df['cleaned'].iloc[0][:80])
```

**Train/Test Split**

```python
# Encode labels
le = LabelEncoder()
df['label'] = le.fit_transform(df['medical_specialty'])

#  Train / Test split (same seed as Day 2)
X_text = df['cleaned']
y      = df['label']

X_train_text, X_test_text, y_train, y_test = train_test_split(
    X_text, y,
    test_size    = 0.20,
    random_state = 42,
    stratify     = y
)

print(f"Train : {len(X_train_text)} samples")
print(f"Test  : {len(X_test_text)} samples")
print(f"Classes: {le.classes_.tolist()}")
```

<h3 style="color: #8B1E3F; margin: 0 0 14px 0; font-size: 18px;">Why LSTM for Text?</h3>

TF-IDF treats text as a bag of words — word order is lost entirely.
LSTM (Long Short-Term Memory) processes text as a sequence,
reading word by word and maintaining a hidden state
that carries context forward.<br><br>

This matters for medical text where order changes meaning:
<code>"no chest pain"</code> vs <code>"chest pain no"</code>
are the same tokens in TF-IDF — but LSTM reads
the negation before the symptom and encodes it correctly.<br><br>

**LSTM**

```python
# ── Tokenize for LSTM (improved) ────────────────────
VOCAB_SIZE  = 20000
MAX_LEN     = 500
EMBED_DIM   = 128
NUM_CLASSES = len(le.classes_)

tokenizer_lstm = Tokenizer(num_words=VOCAB_SIZE, oov_token='<OOV>')
tokenizer_lstm.fit_on_texts(X_train_text)

X_train_seq = pad_sequences(
    tokenizer_lstm.texts_to_sequences(X_train_text),
    maxlen=MAX_LEN, padding='post', truncating='post'
)
X_test_seq = pad_sequences(
    tokenizer_lstm.texts_to_sequences(X_test_text),
    maxlen=MAX_LEN, padding='post', truncating='post'
)

print(f"Vocabulary size : {VOCAB_SIZE}")
print(f"Max sequence len: {MAX_LEN}")
print(f"Train shape     : {X_train_seq.shape}")
print(f"Test shape      : {X_test_seq.shape}")
```

**LSTM Build & Train**

```python
# ── Build improved LSTM ──────────────────────────────
def build_lstm(vocab_size, embed_dim, max_len, num_classes):
    model = models.Sequential([
        layers.Embedding(vocab_size, embed_dim,
                        input_length=max_len, mask_zero=True),
        layers.Bidirectional(layers.LSTM(128, return_sequences=True)),
        layers.Bidirectional(layers.LSTM(64, return_sequences=True)),  # ← True
        layers.GlobalMaxPooling1D(),
        layers.Dropout(0.4),
        layers.Dense(256, activation='relu'),
        layers.Dropout(0.3),
        layers.Dense(num_classes, activation='softmax')
    ])
    return model

lstm_model = build_lstm(VOCAB_SIZE, EMBED_DIM, MAX_LEN, NUM_CLASSES)
lstm_model.compile(
    optimizer = tf.keras.optimizers.Adam(learning_rate=0.001),
    loss      = 'sparse_categorical_crossentropy',
    metrics   = ['accuracy']
)
lstm_model.summary()

#  Class weights
cw = compute_class_weight('balanced',
                           classes=np.unique(y_train), y=y_train)
class_weights = dict(enumerate(cw))

# Train
start = time.time()

cb = [
    callbacks.EarlyStopping(patience=4,
                            restore_best_weights=True,
                            monitor='val_accuracy'),
    callbacks.ReduceLROnPlateau(patience=2, factor=0.5,
                                monitor='val_loss',
                                min_lr=1e-6)
]

history_lstm = lstm_model.fit(
    X_train_seq, y_train,
    validation_split = 0.15,
    epochs           = 20,
    batch_size       = 64,
    class_weight     = class_weights,
    callbacks        = cb,
    verbose          = 1
)

lstm_train_time = time.time() - start
print(f"\n✓ Training complete — {lstm_train_time:.1f}s")
```

**LSTM Evaluate**

```python
# Evaluate
start = time.time()
y_pred_lstm_probs = lstm_model.predict(X_test_seq)
y_pred_lstm       = np.argmax(y_pred_lstm_probs, axis=1)
lstm_infer_time   = time.time() - start

lstm_acc = accuracy_score(y_test, y_pred_lstm)
lstm_f1  = f1_score(y_test, y_pred_lstm, average='macro')

print("LSTM — Results")
print("=" * 45)
print(f"  Accuracy      : {lstm_acc:.4f}")
print(f"  Macro F1      : {lstm_f1:.4f}")
print(f"  Training time : {lstm_train_time:.1f}s")
print(f"  Inference time: {lstm_infer_time:.4f}s")
print()
print(classification_report(
    y_test, y_pred_lstm,
    target_names=le.classes_
))
```

<h3 style="color: #8B1E3F; margin: 0 0 14px 0; font-size: 18px;">DistilBERT vs LSTM</h3>

LSTM reads text sequentially — one word at a time.
DistilBERT uses <strong>self-attention</strong> to look at
all words simultaneously and learn which words

<code>"The patient, despite prior treatment for hypertension,

the relationship between <code>no</code> and <code>cardiac</code>
spans many words. LSTM may lose this over the sequence.
DistilBERT attends to both directly.<br><br>

DistilBERT is a distilled version of BERT —
40% smaller, 60% faster, retains 97% of BERT's performance.
I fine-tune only the classification head today
and freeze the pretrained weights to save time.

**DistilBERT Fine-Tuning**

```python

# Section 6 — DistilBERT Fine-Tuning (PyTorch)

from torch.utils.data import Dataset as TorchDataset, DataLoader
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from torch.optim import AdamW
from torch.nn import CrossEntropyLoss

MODEL_NAME   = 'distilbert-base-uncased'
MAX_LEN_BERT = 128
BATCH_SIZE   = 16
EPOCHS_BERT  = 5

bert_tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

print(f"✓ Tokenizer loaded: {MODEL_NAME}")

#  Custom Dataset
class MedDataset(TorchDataset):
    def __init__(self, texts, labels, tokenizer, max_len):
        self.encodings = tokenizer(
            list(texts),
            max_length  = max_len,
            padding     = 'max_length',
            truncation  = True,
            return_tensors = 'pt'
        )
        self.labels = torch.tensor(list(labels), dtype=torch.long)

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        return {
            'input_ids'      : self.encodings['input_ids'][idx],
            'attention_mask' : self.encodings['attention_mask'][idx],
            'labels'         : self.labels[idx]
        }

print("Tokenizing train set...")
train_dataset_bert = MedDataset(
    X_train_text.values, y_train.values,
    bert_tokenizer, MAX_LEN_BERT
)
print("Tokenizing test set...")
test_dataset_bert = MedDataset(
    X_test_text.values, y_test.values,
    bert_tokenizer, MAX_LEN_BERT
)

train_loader = DataLoader(train_dataset_bert,
                          batch_size=BATCH_SIZE, shuffle=True)
test_loader  = DataLoader(test_dataset_bert,
                          batch_size=BATCH_SIZE)

print(f"✓ Datasets ready — {len(train_dataset_bert)} train · {len(test_dataset_bert)} test")
```

**DistilBERT Train**

```python
# Load model
bert_model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels = NUM_CLASSES
)

# Freeze base
for param in bert_model.distilbert.parameters():
    param.requires_grad = False

# Unfreeze
for layer in bert_model.distilbert.transformer.layer[-2:]:
    for param in layer.parameters():
        param.requires_grad = True

bert_model = bert_model.to(device)

# Optimizer +learning rates
optimizer = AdamW([
    {'params': bert_model.distilbert.transformer.layer[-2:].parameters(),
     'lr': 2e-5},
    {'params': bert_model.classifier.parameters(),
     'lr': 5e-5}
])

loss_fn = CrossEntropyLoss()

# Warmup scheduler
from transformers import get_linear_schedule_with_warmup

EPOCHS_BERT   = 5
total_steps   = len(train_loader) * EPOCHS_BERT
warmup_steps  = int(0.1 * total_steps)

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps   = warmup_steps,
    num_training_steps = total_steps
)

#  Training loop
start = time.time()
best_acc = 0

for epoch in range(EPOCHS_BERT):
    bert_model.train()
    total_loss = 0
    correct    = 0
    total      = 0

    for batch in train_loader:
        input_ids      = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        labels         = batch['labels'].to(device)

        optimizer.zero_grad()
        outputs = bert_model(input_ids=input_ids,
                             attention_mask=attention_mask)
        loss = loss_fn(outputs.logits, labels)
        loss.backward()

        torch.nn.utils.clip_grad_norm_(bert_model.parameters(), 1.0)

        optimizer.step()
        scheduler.step()

        total_loss += loss.item()
        correct    += (outputs.logits.argmax(dim=1) == labels).sum().item()
        total      += labels.size(0)

    acc = correct / total
    print(f"Epoch {epoch+1}/{EPOCHS_BERT} — "
          f"Loss: {total_loss/len(train_loader):.4f} · "
          f"Train Acc: {acc:.4f}")

bert_train_time = time.time() - start

#  Print trainable params
total_params     = sum(p.numel() for p in bert_model.parameters())
trainable_params = sum(p.numel() for p in bert_model.parameters()
                       if p.requires_grad)
print(f"\n✓ Training complete — {bert_train_time:.1f}s")
print(f"  Total params     : {total_params:,}")
print(f"  Trainable params : {trainable_params:,}")
print(f"  Frozen params    : {total_params - trainable_params:,}")
```

**DistilBERT Evaluate**

```python
# Evaluate
bert_model.eval()
all_preds  = []
all_labels = []

start = time.time()
with torch.no_grad():
    for batch in test_loader:
        input_ids      = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        labels         = batch['labels'].to(device)

        outputs = bert_model(input_ids=input_ids,
                             attention_mask=attention_mask)
        preds   = outputs.logits.argmax(dim=1)

        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

bert_infer_time = time.time() - start

y_pred_bert = np.array(all_preds)
y_test_arr  = np.array(all_labels)

bert_acc = accuracy_score(y_test_arr, y_pred_bert)
bert_f1  = f1_score(y_test_arr, y_pred_bert, average='macro')

print("DistilBERT — Results")
print("=" * 45)
print(f"  Accuracy      : {bert_acc:.4f}")
print(f"  Macro F1      : {bert_f1:.4f}")
print(f"  Training time : {bert_train_time:.1f}s")
print(f"  Inference time: {bert_infer_time:.4f}s")
print()
print(classification_report(
    y_test_arr, y_pred_bert,
    target_names=le.classes_
))
```

<strong style="color: #C44569;">The Accuracy vs Macro F1 paradox:</strong><br>
DistilBERT has the highest accuracy (48.4%) but

Gastroenterology, SOAP / Chart, and Urology (recall 0.00).<br><br>

High accuracy on a class-imbalanced dataset is misleading —
Surgery alone is 30% of the test set.
A model that predicts Surgery for every input
achieves 30% accuracy without learning anything useful.
<strong>Macro F1 is the honest metric here</strong>
because it weights all 10 specialties equally.<br><br>

<strong style="color: #C44569;">Why did deep learning underperform here?</strong><br>
Both LSTM and DistilBERT suffer from the same root cause —
<strong>class imbalance without adequate correction.</strong>
The <code>class_weight</code> parameter available in
scikit-learn's Logistic Regression directly penalizes
misclassifying minority classes during training.
PyTorch's training loop used here did not include
a weighted loss function, allowing both models
to exploit the Surgery majority.<br><br>

across all 10 specialties — no class collapses to zero.

but complete failure on 3 specialties.
This is the cost of ignoring class imbalance
in a deep learning training loop.

<strong>If the priority is raw accuracy and resources allow</strong>
— DistilBERT at 48.4% accuracy is the choice.
But it must be retrained with a weighted loss function
to fix the class imbalance problem identified today.
Without that fix, the 48.4% accuracy is misleading —
it is driven by Surgery dominance, not genuine
multi-class understanding.<br><br>

<strong style="color: #C44569;">What today's comparison actually proved:</strong><br>
The most important finding is not which model won —
it is that <strong>class imbalance handling matters more
than model architecture</strong> on this dataset.<br>
TF-IDF + LR with <code>class_weight='balanced'</code>
outperformed BiLSTM and partially-frozen DistilBERT
on the metric that counts (Macro F1)
simply because it was the only model
that explicitly corrected for the imbalance.<br><br>

<strong style="color: #C44569;">Next steps if continuing this project:</strong><br>
1. Retrain DistilBERT with weighted CrossEntropyLoss<br>
2. Filter to top 6 specialties (remove classes with &lt;50 samples)<br>
3. Collect more data for Gastroenterology, SOAP, Urology<br>
4. Try a hierarchical classifier — broad group first, then specialty
