<div style="color: #8B1E3F; font-weight: 700; font-size: 12px; letter-spacing: 1.5px; text-transform: uppercase; margin-bottom: 12px;">
    Why NLP Preprocessing — Today's Focus
</div>
In Week 7, I worked with DistilBERT — a large pretrained Transformer
that handles raw text almost directly through its own built-in tokenizer.<br><br>
Today is different. Before feeding text into any classical model
(TF-IDF, Logistic Regression) or training an LSTM from scratch,
raw text cannot enter the pipeline as-is.<br><br>

<div style="background: white; border-radius: 8px; padding: 14px 18px; margin: 12px 0; font-style: italic; color: #555; border-left: 3px solid #C44569;">
"Patient was NOT responding to treatment. The Dr. administered 500mg of acetaminophen."
</div>

It contains:
<ul style="margin: 12px 0; padding-left: 20px; color: #444;">
  <li style="margin-bottom: 6px;">Punctuation marks — <code>.</code> <code>,</code> <code>(</code> <code>)</code></li>
  <li style="margin-bottom: 6px;">Mixed casing — <code>Dr.</code> vs <code>dr</code></li>
  <li style="margin-bottom: 6px;">Filler words — <code>was</code>, <code>the</code>, <code>of</code></li>
  <li style="margin-bottom: 6px;">Multiple forms of the same word — <code>administered</code>, <code>administering</code>, <code>administers</code></li>
</ul>

All of this is noise to a model that sees words as tokens.
Today's goal is to build a pipeline that removes the noise
while preserving what actually carries meaning.
</div>

<h3 style="color: #8B1E3F; margin: 0 0 14px 0; font-size: 18px;">The Core Problem</h3>
A machine learning model does not read text the way a human does.
It sees <strong style="color: #C44569;">numbers — nothing else.</strong><br><br>
Every word must be converted into a number (or a vector of numbers)
before any model can process it. This conversion is called
<strong>text representation</strong>, and it is covered in Day 2.<br><br>
But there is a problem that comes <em>before</em> representation:
<strong>the same meaning can appear in dozens of different surface forms.</strong>
</div>

different words, different cases, different punctuation.

```python
# ═══════════════════════════════════════════════════
# Section 2 — The preprocessing problem, demonstrated
# ═══════════════════════════════════════════════════

# Three sentences — same meaning, different surface forms
sentences = [
    "The patient is not responding to medication.",
    "Patient NOT responding to medications.",
    "patient not respond to Medication!"
]

# Tokenize each sentence naively (just split on spaces)
print("=" * 55)
print("RAW TOKENS — split on whitespace only")
print("=" * 55)

all_tokens = []
for i, sent in enumerate(sentences, 1):
    tokens = sent.split()
    all_tokens.extend(tokens)
    print(f"\nSentence {i}: {tokens}")

# Unique vocabulary the model would see
raw_vocab = set(all_tokens)
print(f"\nUnique tokens across all 3 sentences : {len(raw_vocab)}")
print(f"Tokens  : {sorted(raw_vocab)}")
print()
print("→ The model sees these as DIFFERENT words:")
print("  'Medication' vs 'medication' vs 'Medication!'")
print("  'medications' vs 'medication'")
print("  'responding' vs 'respond'")
print()
print("→ After full preprocessing, all 3 sentences")
print("  should share the same core vocabulary.")
```

<h3 style="color: #8B1E3F; margin: 0 0 14px 0; font-size: 18px;">What is Tokenization?</h3>

Tokenization is the process of splitting raw text into smaller units called <strong>tokens</strong>.
A token is the basic unit the model works with — think of it as the smallest
meaningful piece of text.<br><br>

Before any model can process language, the text must be broken down.
You cannot feed a full paragraph into a model —
you feed it a sequence of tokens, each converted to a number.<br><br>

<div style="
    background: #fff5f8;

">
"Patient was not responding to treatment."<br>
→ ['Patient', 'was', 'not', 'responding', 'to', 'treatment', '.']
</div>

Each element in that list is one token.
The model sees the list — never the original string.
</div>

**Step 1: Lowercase**

```python
# ═══════════════════════════════════════════════════
# Section 6 — Cleaning & Normalization
# ═══════════════════════════════════════════════════

# Work on the same sample text from Section 5
sample = df['transcription'].iloc[0]

# ── Step 1: Lowercase ────────────────────────────────
step1 = sample.lower()

print("STEP 1 — Lowercase")
print("-" * 50)
print("Before:", sample[:80])
print()
print("After :", step1[:80])
print()

# Count how many tokens changed
before_tokens = set(word_tokenize(sample))
after_tokens  = set(word_tokenize(step1))
print(f"Unique tokens before : {len(before_tokens)}")
print(f"Unique tokens after  : {len(after_tokens)}")
print(f"Tokens eliminated    : {len(before_tokens) - len(after_tokens)}")
```

**Step 2: Remove Punctuation & Numbers**

```python
# ── Step 2: Remove punctuation & numbers ─────────────
import re

step2 = re.sub(r'[^a-z\s]', '', step1)

print("STEP 2 — Remove punctuation & numbers")
print("-" * 50)
print("Before:", step1[:100])
print()
print("After :", step2[:100])
print()

# Tokenize and count
tokens_step2 = word_tokenize(step2)
print(f"Token count after step 2 : {len(tokens_step2)}")
print(f"Unique tokens            : {len(set(tokens_step2))}")
```

**Fix the Regex**

```python
# ── Step 2 Improved — handle hyphens correctly ───────

def clean_text(text):
    # 1. Lowercase
    text = text.lower()
    # 2. Replace hyphens with space (year-old → year old)
    text = text.replace('-', ' ')
    # 3. Remove all non-alphabetic characters
  
    text = re.sub(r'[^a-z0-9\s]', '', text)
    text = re.sub(r'\b\d\b', '', text)
    # 4. Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()
    return text

sample_cleaned = clean_text(sample)

print("STEP 2 IMPROVED — Clean text")
print("-" * 50)
print("Before:", sample[:100])
print()
print("After :", sample_cleaned[:100])
print()

tokens_cleaned = word_tokenize(sample_cleaned)
print(f"Token count  : {len(tokens_cleaned)}")
print(f"Unique tokens: {len(set(tokens_cleaned))}")
print()
print("Top 15 tokens after cleaning:")
freq_cleaned = Counter(tokens_cleaned)
for token, count in freq_cleaned.most_common(15):
    bar = "█" * count
    print(f"  {token:<15} {count:>4}  {bar}")
```

<h3 style="color: #8B1E3F; margin: 0 0 14px 0; font-size: 18px;">What are Stop Words?</h3>

Stop words are the most frequent words in a language —
words that appear in almost every sentence regardless of topic or meaning.
Examples: <code>the</code>, <code>a</code>, <code>is</code>, <code>was</code>,
<code>and</code>, <code>of</code>, <code>to</code>.<br><br>

NLTK provides a built-in list of <strong>179 English stop words</strong>.
The question is not whether they are common — they are.
The question is: <strong>does removing them help or hurt this specific task?</strong><br><br>

The answer depends on the task. For medical specialty classification,
stop words tell the model nothing about which department wrote the report.
<code>she</code> and <code>the</code> appear equally in Surgery and Neurology reports.
Removing them forces the model to focus on what actually differs —
words like <code>cardiac</code>, <code>fracture</code>, <code>biopsy</code>.
</div>

**Stop Words Code**

```python
# ═══════════════════════════════════════════════════
# Section 7 — Stop Words
# ═══════════════════════════════════════════════════

# ── Load NLTK stop words ─────────────────────────────
stop_words = set(stopwords.words('english'))
print(f"NLTK stop words total: {len(stop_words)}")
print()

# ── Show negations that are in the stop list ─────────
negations = {'no', 'not', 'nor', 'never', 'neither',
             'nobody', 'nothing', 'nowhere', 'none'}

in_list = negations & stop_words
print(f"Negation words inside NLTK stop list: {in_list}")
print()

# ── Build custom stop list — remove negations ────────
custom_stops = stop_words - negations
print(f"Custom stop list size (negations removed): {len(custom_stops)}")
```

**Apply Stop Word Removal**

```python
# ── Apply to sample text ─────────────────────────────
tokens_before = word_tokenize(sample_cleaned)
tokens_after  = [t for t in tokens_before if t not in custom_stops]

print(f"Tokens before stop word removal : {len(tokens_before)}")
print(f"Tokens after                    : {len(tokens_after)}")
print(f"Tokens removed                  : {len(tokens_before) - len(tokens_after)}")
print()

# ── Top 15 after stop word removal ───────────────────
freq_after = Counter(tokens_after)
print("Top 15 tokens after stop word removal:")
print("-" * 40)
for token, count in freq_after.most_common(15):
    bar = "█" * count
    print(f"  {token:<15} {count:>4}  {bar}")
print()

# ── Verify negations preserved ───────────────────────
print("Negation check:")
for neg in sorted(negations):
    kept = neg in tokens_after
    status = " preserved" if kept else " removed"
    print(f"  {neg:<10} {status}")
```

**Pipeline Function**

```python
# ═══════════════════════════════════════════════════
# Section 9 — Complete Cleaning Pipeline
# ═══════════════════════════════════════════════════

stemmer      = PorterStemmer()
lemmatizer   = WordNetLemmatizer()
stop_words   = set(stopwords.words('english'))
negations    = {'no', 'not', 'nor', 'never', 'neither',
                'nobody', 'nothing', 'nowhere', 'none'}
custom_stops = stop_words - negations

def get_wordnet_pos(word):
    """Map POS tag to WordNet POS for accurate lemmatization."""
    tag = nltk.pos_tag([word])[0][1][0].upper()
    tag_dict = {
        'J': wordnet.ADJ,
        'N': wordnet.NOUN,
        'V': wordnet.VERB,
        'R': wordnet.ADV
    }
    return tag_dict.get(tag, wordnet.NOUN)

def preprocess(text):
    """
    Full NLP preprocessing pipeline — improved version.
    Changes from original:
    1. Preserves medical numbers (type2, stage3, grade4)
    2. POS-aware lemmatization — nouns stay as nouns
       e.g. allergies → allergy (not allergies)
    """
    # 1. Lowercase
    text = str(text).lower()
    # 2. Replace hyphens with space
    text = text.replace('-', ' ')
    # 3. Keep letters + digits (preserve medical numbers)
    text = re.sub(r'[^a-z0-9\s]', '', text)
    # 4. Remove standalone single digits (noise)
    text = re.sub(r'\b\d\b', '', text)
    # 5. Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()
    # 6. Tokenize
    tokens = word_tokenize(text)
    # 7. Remove stop words (keep negations)
    tokens = [t for t in tokens if t not in custom_stops]
    # 8. POS-aware lemmatization
    tokens = [lemmatizer.lemmatize(t, get_wordnet_pos(t))
              for t in tokens]
    return ' '.join(tokens)

print("Pipeline function defined successfully.")
print()

# Quick test
result = preprocess(df['transcription'].iloc[0])
print("Sample output (first 200 chars):")
print("-" * 50)
print(result[:200])
print()

# Test POS lemmatization improvement
test_words_pos = ['allergies', 'medications', 'diagnoses',
                  'type2', 'stage3', 'grade4']
print("POS Lemmatization test:")
print("-" * 35)
for w in test_words_pos:
    clean = w.lower()
    lemma = lemmatizer.lemmatize(clean, get_wordnet_pos(clean))
    print(f"  {w:<15} → {lemma}")
```

```python
# ── Vocabulary reduction visualization ───────────────
categories = ['Total Tokens\n(millions)', 'Unique Vocabulary']
before_vals = [df['raw_tokens'].sum() / 1_000_000,
               len(set(' '.join(df['transcription'].astype(str)).split())) / 1000]
after_vals  = [df['cleaned_tokens'].sum() / 1_000_000,
               len(set(' '.join(df['cleaned']).split())) / 1000]

x = range(len(categories))
width = 0.35

fig, ax = plt.subplots(figsize=(8, 5))
bars1 = ax.bar([i - width/2 for i in x], before_vals,
               width, label='Before', color='#F8A5C2', edgecolor='white')
bars2 = ax.bar([i + width/2 for i in x], after_vals,
               width, label='After',  color='#C44569', edgecolor='white')

for bar in bars1:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.02,
            f'{bar.get_height():.1f}', ha='center', fontsize=11,
            color='#8B1E3F', fontweight='600')
for bar in bars2:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.02,
            f'{bar.get_height():.1f}', ha='center', fontsize=11,
            color='#8B1E3F', fontweight='600')

ax.set_ylabel('Count (M = millions, K = thousands)', fontsize=11)
ax.set_title('Vocabulary Reduction — Before vs After Pipeline',
             fontsize=13, fontweight='bold', color='#8B1E3F', pad=15)
ax.set_xticks(list(x))
ax.set_xticklabels(categories, fontsize=12)
ax.legend(fontsize=11)
ax.spines[['top', 'right']].set_visible(False)
plt.tight_layout()
plt.show()
```
