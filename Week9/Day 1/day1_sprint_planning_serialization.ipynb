<div style="

<h3 style="color: #8B1E3F; margin: 0 0 14px 0; font-size: 18px;">The Problem</h3>

Training a CNN takes time and GPU compute. Every time the model is needed — in an API, in a web app, in production — retraining from scratch is not an option.<br><br>

<strong>Serialization</strong> solves this: it saves the fully trained model to a file so it can be loaded instantly anywhere, without retraining. The model's weights, architecture, and configuration are all preserved.

<div style="
    background: #fff5f8;
    border-radius: 8px;
    padding: 14px 18px;
    font-family: monospace;
    font-size: 14px;
    color: #555;
    border-left: 3px solid #C44569;
    margin: 16px 0;
">
    Train once → Save → Load anywhere → Predict instantly
</div>

For image models specifically, the preprocessing steps must be applied identically at prediction time:
<ul style="margin: 12px 0; padding-left: 20px; color: #444;">
  <li style="margin-bottom: 6px;">Resize to <code>224×224</code> (same as training)</li>
  <li style="margin-bottom: 6px;">Normalize pixel values to <code>[0, 1]</code> (divide by 255)</li>
  <li style="margin-bottom: 6px;">Expand dims to match batch shape <code>(1, 224, 224, 3)</code></li>
</ul>

These steps will be applied identically in the API and Streamlit app on Day 2 and Day 3.
</div>

<div style="
    background: white;
    border-radius: 12px;
    padding: 24px 28px;
    margin-bottom: 16px;
    border: 1px solid #f2d4de;
    font-size: 15px;
">
<h3 style="color: #8B1E3F; margin: 0 0 16px 0; font-size: 18px;">The Training/Serving Skew Problem</h3>

The most common deployment mistake is saving the model but rewriting the preprocessing by hand in the API. Even a small difference — wrong resize dimensions, wrong normalization range — will silently produce wrong predictions.<br><br>

<table style="width: 100%; border-collapse: collapse; font-size: 14px;">
  <thead>
    <tr style="background: linear-gradient(90deg, #8B1E3F, #C44569);">
      <th style="padding: 10px 14px; color: white; text-align: left;">Approach</th>
      <th style="padding: 10px 14px; color: white; text-align: left;">Risk</th>
    </tr>
  </thead>
  <tbody>
    <tr style="border-bottom: 1px solid #f2d4de; background: #fff5f8;">
      <td style="padding: 10px 14px; color: #C44569; font-weight: 600;">❌ Rewrite preprocessing in the API</td>
      <td style="padding: 10px 14px; color: #555;">Silent errors — wrong predictions with no warning</td>
    </tr>
    <tr>
      <td style="padding: 10px 14px; color: #2d8a4e; font-weight: 600;">✅ Use the same preprocessing code</td>
      <td style="padding: 10px 14px; color: #555;">Guaranteed consistency between training and serving</td>
    </tr>
  </tbody>
</table>

<div style="margin-top: 14px; font-size: 14px; color: #555; line-height: 1.7;">
The rule: <strong>save the model, and keep the preprocessing logic identical</strong> — not reimplemented, identical.
</div>

<h3 style="color: #8B1E3F; margin: 0 0 14px 0; font-size: 18px;">What is MLOps?</h3>

Deployment is where the reproducibility discipline built since Week 1 pays off. Three practices matter now:

```python
AUTOTUNE = tf.data.AUTOTUNE
IMG_SIZE = (112, 112)
BATCH_SIZE = 64

def load_and_preprocess(path, label):
    img = tf.io.read_file(path)
    img = tf.image.decode_jpeg(img, channels=3)
    img = tf.image.resize(img, IMG_SIZE)
    img = img / 255.0
    return img, label

def augment(img, label):
    img = tf.image.random_flip_left_right(img)
    img = tf.image.random_brightness(img, 0.2)
    img = tf.image.random_contrast(img, 0.8, 1.2)
    return img, label

all_paths, all_labels = [], []
class_names = sorted(os.listdir(DATA_DIR))
class_map = {name: i for i, name in enumerate(class_names)}
print("Classes:", class_map)

for class_name in class_names:
    class_dir = os.path.join(DATA_DIR, class_name)
    for fname in os.listdir(class_dir):
        if fname.lower().endswith(('.jpg', '.jpeg', '.png')):
            all_paths.append(os.path.join(class_dir, fname))
            all_labels.append(class_map[class_name])

print(f"Total images: {len(all_paths)}")

idx = np.random.permutation(len(all_paths))
split = int(len(idx) * 0.8)
train_idx, val_idx = idx[:split], idx[split:]

train_paths  = [all_paths[i] for i in train_idx]
train_labels = [all_labels[i] for i in train_idx]
val_paths    = [all_paths[i] for i in val_idx]
val_labels   = [all_labels[i] for i in val_idx]

train_ds = (tf.data.Dataset.from_tensor_slices((train_paths, train_labels))
            .map(load_and_preprocess, num_parallel_calls=AUTOTUNE)
            .map(augment, num_parallel_calls=AUTOTUNE)
            .shuffle(1000)
            .batch(BATCH_SIZE)
            .prefetch(AUTOTUNE))

val_ds = (tf.data.Dataset.from_tensor_slices((val_paths, val_labels))
          .map(load_and_preprocess, num_parallel_calls=AUTOTUNE)
          .batch(BATCH_SIZE)
          .prefetch(AUTOTUNE))

print(f"Train batches: {len(train_ds)}")
print(f"Val batches  : {len(val_ds)}")
print("Datasets ready ✓")
```

    border-bottom: 1px solid #f2d4de;
">
    Build Model
</div>

```python

# CNN + Data Augmentation — same architecture as Week 7


def build_model():
    model = keras.Sequential([
        layers.Conv2D(32, (3,3), activation='relu', padding='same',
                      input_shape=(112, 112, 3)),
        layers.BatchNormalization(),
        layers.MaxPooling2D(2, 2),

        layers.Conv2D(64, (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D(2, 2),

        layers.Conv2D(128, (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D(2, 2),

        layers.Conv2D(256, (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D(2, 2),

        layers.GlobalAveragePooling2D(),
        layers.Dense(256, activation='relu'),
        layers.Dropout(0.5),
        layers.Dense(1, activation='sigmoid')
    ])

    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=0.001),
        loss='binary_crossentropy',
        metrics=['accuracy', keras.metrics.AUC(name='auc')]
    )
    return model

model = build_model()
model.summary()
```

    border-bottom: 1px solid #f2d4de;
">
    Train Model
</div>

```python

callbacks = [
    keras.callbacks.EarlyStopping(
        monitor='val_auc',
        patience=5,
        mode='max',
        restore_best_weights=True,
        verbose=1
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor='val_auc',
        factor=0.5,
        patience=3,
        mode='max',
        verbose=1
    )
]

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=30,
    callbacks=callbacks,
    verbose=1
)

print("\nTraining complete ✓")
print(f"Best val_auc : {max(history.history['val_auc']):.4f}")
print(f"Best val_acc : {max(history.history['val_accuracy']):.4f}")
```
