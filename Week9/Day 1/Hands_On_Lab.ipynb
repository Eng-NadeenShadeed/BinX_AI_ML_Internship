I retrained the CNN + Augmentation model on the ISIC Melanoma dataset using a <code>tf.data</code> pipeline for faster GPU utilization. The model was saved using Keras's native <code>.keras</code> format, which preserves the full architecture, weights, and training configuration in a single file.
</div>

**Load Data**

```python

# Step 2 — Build tf.data pipeline & retrain


AUTOTUNE   = tf.data.AUTOTUNE
IMG_SIZE   = (112, 112)
BATCH_SIZE = 64

def load_and_preprocess(path, label):
    img = tf.io.read_file(path)
    img = tf.image.decode_jpeg(img, channels=3)
    img = tf.image.resize(img, IMG_SIZE)
    img = img / 255.0
    return img, label

def augment(img, label):
    img = tf.image.random_flip_left_right(img)
    img = tf.image.random_brightness(img, 0.2)
    img = tf.image.random_contrast(img, 0.8, 1.2)
    return img, label

all_paths, all_labels = [], []
class_names = sorted(os.listdir(DATA_DIR))
class_map   = {name: i for i, name in enumerate(class_names)}

for class_name in class_names:
    class_dir = os.path.join(DATA_DIR, class_name)
    for fname in os.listdir(class_dir):
        if fname.lower().endswith(('.jpg', '.jpeg', '.png')):
            all_paths.append(os.path.join(class_dir, fname))
            all_labels.append(class_map[class_name])

idx        = np.random.permutation(len(all_paths))
split      = int(len(idx) * 0.8)
train_idx, val_idx = idx[:split], idx[split:]

train_paths  = [all_paths[i] for i in train_idx]
train_labels = [all_labels[i] for i in train_idx]
val_paths    = [all_paths[i] for i in val_idx]
val_labels   = [all_labels[i] for i in val_idx]

train_ds = (tf.data.Dataset.from_tensor_slices((train_paths, train_labels))
            .map(load_and_preprocess, num_parallel_calls=AUTOTUNE)
            .map(augment, num_parallel_calls=AUTOTUNE)
            .shuffle(1000)
            .batch(BATCH_SIZE)
            .prefetch(AUTOTUNE))

val_ds = (tf.data.Dataset.from_tensor_slices((val_paths, val_labels))
          .map(load_and_preprocess, num_parallel_calls=AUTOTUNE)
          .batch(BATCH_SIZE)
          .prefetch(AUTOTUNE))

print(f"Classes    : {class_map}")
print(f"Train imgs : {len(train_paths)}")
print(f"Val imgs   : {len(val_paths)}")
print("Pipeline ready ✓")
```

**Build & Train**

```python

# Build & Train


def build_model():
    model = keras.Sequential([
        layers.Conv2D(32, (3,3), activation='relu', padding='same',
                      input_shape=(112, 112, 3)),
        layers.BatchNormalization(),
        layers.MaxPooling2D(2, 2),

        layers.Conv2D(64, (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D(2, 2),

        layers.Conv2D(128, (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D(2, 2),

        layers.Conv2D(256, (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D(2, 2),

        layers.GlobalAveragePooling2D(),
        layers.Dense(256, activation='relu'),
        layers.Dropout(0.5),
        layers.Dense(1, activation='sigmoid')
    ])

    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=0.001),
        loss='binary_crossentropy',
        metrics=['accuracy', keras.metrics.AUC(name='auc')]
    )
    return model

model = build_model()

callbacks = [
    keras.callbacks.EarlyStopping(
        monitor='val_auc',
        patience=5,
        mode='max',
        restore_best_weights=True,
        verbose=1
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor='val_auc',
        factor=0.5,
        patience=3,
        mode='max',
        verbose=1
    )
]

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=30,
    callbacks=callbacks,
    verbose=1
)

print("\nTraining complete ✓")
print(f"Best val_auc : {max(history.history['val_auc']):.4f}")
print(f"Best val_acc : {max(history.history['val_accuracy']):.4f}")
```

```python

# Step 2 — Save the model


MODEL_PATH = os.path.join(SAVE_DIR, 'melanoma_cnn_lab.keras')
model.save(MODEL_PATH)

print(f"Model saved ✓")
print(f"Path: {MODEL_PATH}")
print(f"Size: {os.path.getsize(MODEL_PATH) / 1024 / 1024:.1f} MB")
```

```python

# Verify on 10 samples

class_names = ['Benign', 'Malignant']
correct = 0
total   = 10

print(f"{'Image':<15} {'True':<12} {'Predicted':<12} {'Conf':<8} {'Result'}")
print("─" * 60)

for i in range(total):
    sample_path  = val_paths[i]
    sample_label = val_labels[i]

    img = tf.io.read_file(sample_path)
    img = tf.image.decode_jpeg(img, channels=3)
    img = tf.image.resize(img, (112, 112))
    img = img / 255.0
    img = tf.expand_dims(img, axis=0)

    pred_prob  = loaded_model.predict(img, verbose=0)[0][0]
    pred_class = 1 if pred_prob >= 0.5 else 0
    is_correct = pred_class == sample_label

    if is_correct:
        correct += 1

    print(f"{sample_path.split('/')[-1]:<15} {class_names[sample_label]:<12} "
          f"{class_names[pred_class]:<12} {pred_prob:.4f}   "
          f"{'✅' if is_correct else '❌'}")

print("─" * 60)
print(f"Accuracy on 10 samples: {correct}/{total} ({correct/total*100:.0f}%)")
```

```python

# Step 4 — Freeze requirements.txt


import subprocess

req_path = os.path.join(SAVE_DIR, '../requirements.txt')
req_path = os.path.normpath(req_path)

result = subprocess.run(['pip', 'freeze'], capture_output=True, text=True)

keywords = ['tensorflow', 'keras', 'numpy', 'pillow', 'matplotlib']
lines = [l for l in result.stdout.split('\n')
         if any(k in l.lower() for k in keywords)]

with open(req_path, 'w') as f:
    f.write('\n'.join(lines))

print("requirements.txt saved ✓")
print(f"Path: {req_path}")
print("\nContents:")
for l in lines:
    print(f"  {l}")
```
