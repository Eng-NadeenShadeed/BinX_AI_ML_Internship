<h3 style="color: #8B1E3F; margin: 0 0 14px 0; font-size: 18px;">
From Local Application to Public Application
</h3>

<b>Deployment</b> is the process of taking an application from a development environment and making it available in an environment where users can access and use it.

During development, an application may run only on the developer's computer. For example, the Streamlit dashboard created on Day 3 runs locally through <code>localhost:8501</code>.

Public deployment moves the application to a hosting environment connected to the internet. The hosting platform runs the application and provides a public URL that users can open from their own devices.

<h3 style="color: #8B1E3F; margin: 0 0 14px 0; font-size: 18px;">
Why Do File Names and Structure Matter?
</h3>

Deployment systems execute commands relative to the files available in the deployed project.

If the application expects a file at a specific location, changing the project structure without updating the corresponding path can cause the application to fail.

For example, my application uses a relative path to locate the serialized CNN model.

Therefore, the relationship between:

<ul>
<li>the application file,</li>
<li>the model location, and</li>
<li>the path used in the Python code</li>
</ul>

<h3 style="color: #8B1E3F; margin: 0 0 14px 0; font-size: 18px;">
What Is a Port?
</h3>

A <b>port</b> is a logical communication endpoint through which a network service can receive connections.

When I ran the Streamlit application locally, it was available through:

<pre style="
    background: #2d2528;
    color: #f8f5f2;
    padding: 18px;
    border-radius: 8px;

    font-size: 14px;
">
http://localhost:8501
</pre>

Here, <code>8501</code> is the port used by Streamlit by default in the local setup.

In a hosted environment, the platform manages how external web traffic reaches the running application. The application must therefore be started in a way that is compatible with the hosting platform's networking requirements.

</ul>

<h3 style="color: #8B1E3F; margin: 0 0 18px 0; font-size: 18px;">
Key Takeaways
</h3>

<ul>
<li><b>Deployment</b> moves an application from a local environment to a hosted environment.</li>

<li><b>Hosting platforms</b> provide the infrastructure needed to build and run applications remotely.</li>

<li><b>requirements.txt</b> defines the Python dependencies required by the deployed application.</li>

<li><b>Reproducibility</b> makes it easier to recreate the application environment consistently.</li>

<li><b>Relative paths</b> help applications locate project files across different environments.</li>

<li><b>Build</b> prepares the environment, while <b>runtime</b> executes the application.</li>

<li><b>Deployment logs</b> help identify and diagnose build and runtime problems.</li>

<li><b>Local testing</b> is necessary but not sufficient; the live application must also be tested.</li>

<li>The deployed application uses the existing serialized CNN rather than retraining the model.</li>

<li>Deployment connects model development with practical <b>MLOps</b> workflows.</li>
</ul>
