<h3 style="color: #8B1E3F; margin: 0 0 14px 0; font-size: 18px;">From Notebook to Service</h3>

A trained model sitting in a notebook helps no one outside that notebook. FastAPI turns it into a service — a URL that any application can call to get a prediction.<br><br>

FastAPI is a modern Python framework for building REST APIs. It handles the HTTP layer: receiving requests, routing them to the right function, and returning responses. The model stays the same; FastAPI is just the layer that makes it accessible over the web.<br><br>

<h3 style="color: #8B1E3F; margin: 0 0 14px 0; font-size: 18px;">Why Validate Input?</h3>

A prediction API is only as reliable as the data it receives. Without validation, a badly formed request — a missing field, a string where a number is expected, an empty file — will crash the model with a confusing error.<br><br>

Pydantic solves this by defining the expected input as a Python class. FastAPI uses that class to automatically validate every incoming request before it reaches the model. If the input is invalid, FastAPI rejects it with a clear error message — the model never sees bad data.<br><br>

app = FastAPI(title=<span style="color: #a6e3a1;">"Melanoma Classifier API"</span>)<br><br>

<span style="color: #6c7086;"># Load model once at startup</span><br>

<span style="color: #89b4fa;">@app.get</span>(<span style="color: #a6e3a1;">"/"</span>)<br>
<span style="color: #cba6f7;">def</span> <span style="color: #89dceb;">root</span>():<br>
&nbsp;&nbsp;&nbsp;&nbsp;<span style="color: #cba6f7;">return</span> {<span style="color: #a6e3a1;">"message"</span>: <span style="color: #a6e3a1;">"Melanoma Classifier API is running"</span>}<br><br>

<span style="color: #89b4fa;">@app.post</span>(<span style="color: #a6e3a1;">"/predict"</span>)<br>
<span style="color: #cba6f7;">async def</span> <span style="color: #89dceb;">predict</span>(file: UploadFile = File(...)):<br>
&nbsp;&nbsp;&nbsp;&nbsp;<span style="color: #6c7086;"># Validate file type</span><br>
&nbsp;&nbsp;&nbsp;&nbsp;<span style="color: #cba6f7;">if</span> file.content_type <span style="color: #cba6f7;">not in</span> [<span style="color: #a6e3a1;">"image/jpeg"</span>, <span style="color: #a6e3a1;">"image/png"</span>]:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span style="color: #cba6f7;">raise</span> HTTPException(status_code=<span style="color: #fab387;">400</span>,<br>

&nbsp;&nbsp;&nbsp;&nbsp;<span style="color: #6c7086;"># Predict</span><br>

&nbsp;&nbsp;&nbsp;&nbsp;pred_class = <span style="color: #a6e3a1;">"Malignant"</span> <span style="color: #cba6f7;">if</span> pred_prob >= <span style="color: #fab387;">0.5</span> <span style="color: #cba6f7;">else</span> <span style="color: #a6e3a1;">"Benign"</span><br>
&nbsp;&nbsp;&nbsp;&nbsp;confidence = pred_prob <span style="color: #cba6f7;">if</span> pred_prob >= <span style="color: #fab387;">0.5</span> <span style="color: #cba6f7;">else</span> <span style="color: #fab387;">1</span> - pred_prob<br><br>
&nbsp;&nbsp;&nbsp;&nbsp;<span style="color: #cba6f7;">return</span> {<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span style="color: #a6e3a1;">"prediction"</span>: pred_class,<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span style="color: #a6e3a1;">"confidence"</span>: <span style="color: #fab387;">round</span>(confidence, <span style="color: #fab387;">4</span>),<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span style="color: #a6e3a1;">"filename"</span>: file.filename<br>
&nbsp;&nbsp;&nbsp;&nbsp;}
