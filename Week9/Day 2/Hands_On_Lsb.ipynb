<h3 style="

    font-size: 18px;
    font-weight: 600;
">
    Prediction Logic
</h3>

<p>
The CNN produces a probability between 0 and 1 using its final sigmoid output.
A threshold of <code>0.5</code> is used to convert this probability into the predicted class.
</p>

<div style="
    background: #2b2729;
    color: #f3ede6;
    padding: 16px 18px;
    border-radius: 7px;
    font-family: monospace;
    font-size: 13px;
    line-height: 1.7;
">
pred_class = "Malignant" if pred_prob &gt;= 0.5 else "Benign"<br><br>
confidence = pred_prob if pred_prob &gt;= 0.5 else 1 - pred_prob
</div>

<p>
For example, if the model outputs <code>0.91</code>, the prediction is <strong>Malignant</strong> with a confidence of <code>0.91</code>.
If the model outputs <code>0.12</code>, the prediction is <strong>Benign</strong> with a confidence of <code>0.88</code>.
</p>

<div style="background: #faf7f2; padding: 24px; border-radius: 14px; border: 1px solid #eee5dc; color: #4a4542;">

<h3 style="color: #8B1E3F; margin-top: 0;">End-to-End Prediction Flow</h3>

<p>
The FastAPI application connects the uploaded image to the trained CNN through a sequence of preprocessing and prediction steps.
</p>

<div style="background: #f3ede6; padding: 18px; border-radius: 10px; line-height: 2; text-align: center;">

<b>Image Upload</b>
<br>↓<br>

<b>UploadFile</b>
<br>↓<br>

<b>Validate File Type</b>
<br>JPEG / PNG only
<br>↓<br>

<b>Read Image</b>
<br>↓<br>

<b>Convert to RGB</b>
<br>↓<br>

<b>Resize to 112 × 112</b>
<br>↓<br>

<b>Normalize Pixel Values</b>
<br><code>/ 255.0</code>
<br>↓<br>

<b>Add Batch Dimension</b>
<br>Shape: <code>(1, 112, 112, 3)</code>
<br>↓<br>

<b>CNN Model</b>
<br>↓<br>

<b>Prediction Probability</b>
<br>↓<br>

<b>Class + Confidence</b>
<br>↓<br>

<b>JSON Response</b>

</div>

<p style="margin-bottom: 0;">
The preprocessing used during inference must match the preprocessing used during training so that the CNN receives data in the expected format.
</p>

</div>

<div style="background: linear-gradient(90deg, #8B1E3F 0%, #C44569 100%); padding: 14px 24px; border-radius: 10px; margin: 28px 0 16px 0; display: flex; align-items: center; gap: 14px;">
    <span style="background: rgba(255,255,255,0.18); color: white; width: 34px; height: 34px; border-radius: 50%; display: flex; align-items: center; justify-content: center; font-weight: bold;">11</span>
    <h2 style="color: white; margin: 0;">Key Takeaways</h2>
</div>

<div style="background: #faf7f2; padding: 24px; border-radius: 14px; border: 1px solid #eee5dc; color: #4a4542;">

<h3 style="color: #8B1E3F; margin-top: 0;">What I Learned and Implemented</h3>

<ul style="line-height: 1.9;">

<li>
<b>FastAPI</b> can expose a trained machine learning model through a REST API.
</li>

<li>
The serialized <b>Keras CNN model</b> is loaded once when the application starts.
</li>

<li>
The inference preprocessing matches the training input format:
<b>RGB → 112×112 → NumPy array → /255.0 → batch dimension</b>.
</li>

<li>
The <b>/predict</b> endpoint accepts image uploads and returns a JSON prediction.
</li>

<li>
The API supports <b>JPEG and PNG</b> images and rejects unsupported file types.
</li>

<li>
Invalid input is handled cleanly using an HTTP <b>400 Bad Request</b> response.
</li>

<li>
FastAPI's automatic <b>Swagger UI</b> at <code>/docs</code> provides an interactive way to test the endpoint locally.
</li>

<li>
The prediction response contains the predicted class, confidence for that predicted class, and the original filename.
</li>

</ul>
