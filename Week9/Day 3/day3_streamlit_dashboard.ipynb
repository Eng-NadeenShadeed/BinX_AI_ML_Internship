<h3 style="color: #8B1E3F; margin-top: 0;">
From Model to User Interface
</h3>

<p>
During model development, we usually interact with a machine learning model
through Python code. However, a non-technical user should not need to open
a notebook or write Python code just to use the model.
</p>

<p>
<strong>Streamlit</strong> is a Python framework that allows us to build
interactive web applications directly from Python.
It provides ready-made UI components such as file uploaders, buttons,
text fields, metrics, and charts.
</p>

<p>
For our melanoma project, Streamlit will provide the
<strong>human-facing interface</strong>.
A user can upload an ISIC skin lesion image, click a button,
and receive the CNN prediction without interacting with Python code.
</p>

</div>

    background: #fff5f8;

<h3 style="color: #8B1E3F; margin-top: 0;">
A Streamlit App Is a Python Script
</h3>

<p>
A Streamlit application is usually a normal Python file.
Instead of building a separate frontend with HTML, CSS, and JavaScript,
we use Streamlit's Python commands to create the interface.
</p>

<p>
For example:
</p>

<pre style="
    background: #282828;
    color: #f5f5f5;
    padding: 18px;
    border-radius: 8px;
    overflow-x: auto;
"><code>import streamlit as st

st.title("Melanoma Classifier")

name = st.text_input("Enter your name")

if name:
    st.write(f"Hello, {name}!")</code></pre>

<p>
The Streamlit commands become visible interface elements in the browser.
</p>

<h3 style="color: #8B1E3F; margin-top: 0;">
The Streamlit Execution Model
</h3>

<p>
One important concept in Streamlit is that the Python script is re-run
from top to bottom whenever the user interacts with the application.
</p>

<p>
For example, clicking a button or uploading a new image can trigger
another execution of the script.
</p>

<p>
This behavior makes Streamlit simple to use, but it also means that
expensive operations such as loading a large machine learning model
should be handled carefully.
</p>

</div>

<h3 style="color: #8B1E3F; margin-top: 0;">
What Are Streamlit Widgets?
</h3>

<p>
<strong>Widgets</strong> are interactive UI elements that allow users
to provide input or interact with a Streamlit application.
</p>

<p>
The appropriate widget depends on the type of input required by the
machine learning model.
</p>

<p>
For example, a numerical prediction system may use sliders or number
inputs, while a text classification system may use a text box.
An image classification system needs a way to upload an image.
</p>

<h3 style="color: #8B1E3F; margin-top: 0;">
Why Do We Use st.file_uploader()?
</h3>

<p>
Our CNN does not expect tabular values such as age, blood pressure,
or cholesterol. It expects an <strong>image tensor</strong>.
</p>

<p>
Therefore, the most appropriate input widget for our application is
<strong>st.file_uploader()</strong>.
</p>

<p>
The user uploads a JPG, JPEG, or PNG image. The application then:
</p>

<ol>
<li>Reads the uploaded image.</li>
<li>Converts it to RGB.</li>
<li>Resizes it to <strong>112 × 112</strong>.</li>
<li>Normalizes pixel values by dividing by <strong>255.0</strong>.</li>
<li>Adds the batch dimension.</li>
<li>Sends the image to the CNN.</li>
</ol>

<h3 style="color: #8B1E3F; margin-top: 0;">
Streamlit Re-runs the Script
</h3>

<p>
Unlike a traditional web application where different parts of the
frontend may update independently, Streamlit normally re-executes
the Python script from top to bottom when the user interacts with
the application.
</p>

<p>
For example, when the user uploads an image or clicks the
<strong>Predict</strong> button, Streamlit can trigger a new execution
of the script.
</p>

<p>
This execution model makes Streamlit very simple for Python developers,
but it also means that we need to think carefully about operations
that are expensive or time-consuming.
</p>

</div>

<div style="
    background: #fff5f8;
    border-left: 5px solid #C44569;
    border-radius: 12px;

">

<h3 style="color: #8B1E3F; margin-top: 0;">
Why Can Re-running Become a Problem?
</h3>

<p>
Our CNN is stored as a serialized Keras model in:
</p>

<pre style="
    background: #282828;
    color: #f5f5f5;
    padding: 16px;
    border-radius: 8px;
"><code>Day 1/models/melanoma_cnn.keras</code></pre>

<p>
Loading this model requires reading the model file and reconstructing
the neural network in memory.
</p>

<p>
If the model is loaded every time Streamlit re-runs the script,
the application performs unnecessary work.
</p>

<p>
This can make the application slower and can waste computational
resources.
</p>

</div>

<div style="
    background: #faf7f2;
    border: 1px solid #eee5dc;
    border-radius: 12px;
    padding: 26px 30px;
    margin: 14px 0;
    color: #3F3A38;
">

<h3 style="color: #8B1E3F; margin-top: 0;">
Caching in Streamlit
</h3>

<p>
Streamlit provides caching mechanisms that allow expensive resources
to be reused instead of recreated on every script execution.
</p>

<p>
For machine learning models, the appropriate decorator is:
</p>

<pre style="
    background: #282828;
    color: #f5f5f5;
    padding: 16px;
    border-radius: 8px;
"><code>@st.cache_resource</code></pre>

<p>
We can use it with a model-loading function so that the model is
loaded once and then reused across Streamlit script re-runs.
</p>

</div>

<div style="
    background: #fff5f8;
    border-left: 5px solid #C44569;
    border-radius: 12px;
    padding: 24px 28px;
    margin: 18px 0;
    color: #3F3A38;
">

<h3 style="color: #8B1E3F; margin-top: 0;">
Loading the CNN Efficiently
</h3>

<p>
For our project, the model-loading function can be written as:
</p>

<pre style="
    background: #282828;
    color: #f5f5f5;
    padding: 18px;
    border-radius: 8px;
    overflow-x: auto;
"><code>@st.cache_resource
def load_model():
    return tf.keras.models.load_model(MODEL_PATH)</code></pre>

<p>
The first time the application needs the model, Streamlit loads it.
On subsequent script re-runs, Streamlit can reuse the cached resource
instead of loading the model from disk again.
</p>

</div>

<div style="
    background: #faf7f2;
    border: 1px solid #eee5dc;
    border-radius: 12px;
    padding: 26px 30px;

<h3 style="color: #8B1E3F; margin-top: 0;">
From Serialized Model to Streamlit Application
</h3>

<p>
In Day 1, we serialized our trained CNN using the Keras
<strong>.keras</strong> format.

<h3 style="color: #8B1E3F; margin-top: 0;">
What Makes an ML Demo Usable?
</h3>

<p>
A machine learning demo should not only produce a prediction.
It should also make the interaction clear and easy for a first-time user.
</p>

<p>
A good demo should provide:
</p>

<ul>
<li><strong>Clear purpose:</strong> The user should immediately understand what the application does.</li>
<li><strong>Simple input:</strong> The required input should be obvious.</li>
<li><strong>Clear action:</strong> The user should know what to click or do next.</li>
<li><strong>Prominent result:</strong> The prediction should be easy to notice.</li>
<li><strong>Supporting information:</strong> Confidence or probabilities can help explain the output.</li>
<li><strong>Useful feedback:</strong> The application should respond clearly when an input is missing or invalid.</li>
</ul>

</div>

<li>
<strong>Image preview:</strong> Shows the uploaded image before prediction.

<li>

</li>

<li>
<strong>Prediction result:</strong> Displays Benign or Malignant prominently.

<li>

<h3 style="color: #8B1E3F; margin-top: 0;">
Making the Prediction Easier to Understand
</h3>

<p>
A classification result is more informative when the user can also see
the model's predicted probabilities.
</p>

<p>
Our CNN uses a <strong>sigmoid</strong> output, which produces a value
between 0 and 1.
For our binary classification task, this value represents the model's
estimated probability for the <strong>Malignant</strong> class.
</p>

<p>
The probability for the Benign class can therefore be calculated as:
</p>

<pre style="
    background: #282828;
    color: #f5f5f5;
    padding: 16px;
    border-radius: 8px;
"><code>Benign Probability = 1 - Malignant Probability</code></pre>

</div>

<div style="
    background: #fff5f8;

<h3 style="color: #8B1E3F; margin-top: 0;">
Prediction Probability and Confidence
</h3>

<p>
The raw sigmoid output represents the predicted probability of the
<strong>Malignant</strong> class.
</p>

<p>
However, the confidence displayed by our application corresponds to
the probability of the <strong>predicted class</strong>.
</p>

<pre style="
    background: #282828;
    color: #f5f5f5;
    padding: 18px;
    border-radius: 8px;
"><code>If prediction = Malignant:
confidence = malignant probability

If prediction = Benign:
confidence = benign probability
            = 1 - malignant probability</code></pre>

<h3 style="color: #8B1E3F; margin-top: 0;">
What I Learned Today
</h3>

<ul>
<li>
<strong>Streamlit</strong> allows us to build interactive web applications
using Python without building a separate frontend.
</li>

<li>
<strong>Widgets</strong> provide interactive inputs for users.
For our image classification project, <code>st.file_uploader()</code>
is the appropriate input widget.
</li>

<li>
Streamlit normally <strong>re-runs the Python script</strong> when the
user interacts with the application.
</li>

<li>
<strong><code>@st.cache_resource</code></strong> can be used to cache
the loaded CNN model and avoid unnecessary model loading.
</li>

<li>
The uploaded image must follow the same <strong>preprocessing pipeline</strong>
used during training: RGB, 112 × 112, normalization, and batch dimension.
</li>

<li>
A usable ML demo should have a <strong>clear input, clear action,
prominent result, and useful supporting information</strong>.
</li>

<li>
<strong>FastAPI</strong> provides a programmatic API, while
<strong>Streamlit</strong> provides a human-facing interface.
</li>

<li>
The final goal is to move the trained CNN from a saved model artifact
to an application that users can interact with.
</li>

</ul>
