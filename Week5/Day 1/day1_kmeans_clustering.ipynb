## 2. Supervised vs Unsupervised — The Full Picture

|                | Supervised                 | Unsupervised                                                    |
| -------------- | -------------------------- | --------------------------------------------------------------- |
| **Data**       | X and y                    | X only                                                          |
| **Goal**       | Predict a known target     | Discover hidden structure                                       |
| **Examples**   | Regression, classification | Clustering, dimensionality reduction, association rule learning |
| **Evaluation** | Compare to true labels     | Internal metrics + judgment                                     |
| **Output**     | Predictions                | Groups, components, rules, or anomalies                         |

In **supervised learning**, the model learns from examples where the correct target is already known. The goal is to learn a mapping between the input features `X` and the target `y`, then apply that mapping to make predictions on new data.

In **unsupervised learning**, there is no predefined target or label. The algorithm receives only the features `X` and searches for patterns, relationships, or hidden structure within the data.

A typical unsupervised learning workflow looks like this:

```text
Unlabeled Data (X)
        ↓
Choose an Unsupervised Algorithm
        ↓
Fit the Algorithm on X
        ↓
Discover Structure or Patterns
        ↓
Interpret and Use the Results
```

Common unsupervised learning tasks include:

* **Clustering** — grouping similar data points
* **Dimensionality reduction** — representing high-dimensional data using fewer dimensions
* **Association rule learning** — discovering relationships between items or events
* **Anomaly detection** — identifying unusual observations

**When is unsupervised learning useful?**

* You have data but no labels (labeling is expensive or impossible)
* You want to understand the natural structure of your data
* You're doing customer segmentation without pre-defined groups
* You want to compress or visualize high-dimensional data
* You're detecting unusual observations without labeled anomaly cases

An important point is that the algorithm does not automatically know what a discovered group means. After finding the structure, I must **interpret the results** using the original features and domain knowledge.

For example, K-Means might discover four customer clusters, but the labels — 0, 1, 2, 3 — carry no meaning on their own. That meaning comes from analyzing the feature profiles of each cluster.

## 3. What Is Clustering?

Clustering is an unsupervised learning task that groups data points so that points within the same group are more similar to each other, while points in different groups are more dissimilar.

No one pre-defines the groups or provides the correct cluster labels. The algorithm discovers groups by looking for patterns and similarities in the feature space, often using distance between data points.

**Real-world examples:**

```text
Customer data → who are my natural customer segments?
Medical data  → which patients have similar risk profiles?
Housing data  → which properties or neighborhoods have similar characteristics?
```

The key insight is:

> **The groups are discovered, not imposed.**

You do not tell the algorithm which observations belong to which group. Instead, the algorithm uses the available features to find groups of observations that are similar according to the chosen similarity or distance measure.

For example, after clustering customer data, it's possible to discover a group with a high percentage of smokers — even though the algorithm was never told about smoking. That pattern only becomes clear when I interpret the feature profiles of each cluster.

### A Bigger Picture of Clustering

Clustering methods use different ideas to define what makes data points belong together:

```text
Clustering
│
├── Centroid-based
│     └── K-Means              ← today
│
├── Density-based
│     └── DBSCAN               ← Day 2
│
├── Connectivity-based
│     └── Hierarchical         ← Day 2
│
└── Distribution-based
      └── Gaussian Mixture Models
```

K-Means is a **centroid-based** and **hard clustering** algorithm: each observation is assigned to exactly one cluster, and each cluster is represented by a centroid.

Other clustering approaches can use different ideas, such as density, connectivity, or probabilistic membership.

For K-Means, the number of clusters `k` must be chosen in advance. This creates an important question:

> **How do I choose a good value of `k`?**

The **Elbow Method** and the **Silhouette Score** are the two tools I'll use to answer this.

## 4. K-Means — How It Works

K-Means is the most widely used clustering algorithm.
It partitions data into k clusters by repeating three steps
until the cluster assignments stop changing:

**Step 1 — Initialize:**
Place k centroids at random positions in the feature space.

**Step 2 — Assign:**
Assign each data point to the nearest centroid.

**Step 3 — Update:**
Move each centroid to the mean position of its assigned points.

**Repeat steps 2–3 until convergence.**

```text
Iteration 1:
  Random centroids → assign points → move centroids

Iteration 2:
  New centroids → reassign points → move centroids

...

Convergence:
  Centroids stop moving → assignment is stable
```

---

**What is a centroid?**

A centroid is the center of a cluster. For K-Means, the centroid
is the mean position of all points assigned to that cluster.

For example, if one feature has the values 2, 4, 6:

```text
centroid = (2 + 4 + 6) / 3 = 4
```

The name **K-Means** comes from:
- **K** → the number of clusters
- **Means** → cluster centers are calculated using means

---

**What K-Means optimizes:**

K-Means minimizes **inertia** — the total within-cluster
sum of squared distances from each point to its centroid.

```text
Low inertia  = compact clusters = points close to their centroid
High inertia = spread-out clusters = points far from their centroid

Inertia = Σ (distance from each point to its centroid)²
```

As k increases, inertia always decreases — more centroids give
the model more flexibility. However, simply choosing the largest
possible k is not useful. If every point had its own cluster,
inertia could reach zero, but the clustering would not provide
meaningful structure.

---

**`n_init=10`**

K-Means can converge to a poor result depending on where the
initial centroids happen to be placed. `n_init=10` means the
algorithm runs 10 times with different random starting centroids
and keeps the result with the lowest inertia.

This reduces the risk of settling on a bad local solution
due to unlucky initialization.

---

**One important limitation:**

K-Means requires you to choose k in advance.
There is no automatic way to determine k — which is why
I'll use the elbow method and the silhouette score.


## 5. Why Scaling Is Required Before Clustering

K-Means uses **distance** to assign data points to their nearest
centroid. Because distance is sensitive to the scale of each feature,
features with much larger numerical ranges can dominate the calculation.

For the `insurance.csv` dataset:

```text
age:      18 – 64
bmi:      approximately 16 – 53
children: 0 – 5
charges:  approximately 1,122 – 63,770
```

The range of `charges` is much larger than the others.
Without scaling, `charges` would dominate every distance calculation —
features like `age`, `bmi`, and `children` would have almost no influence.

For example, when calculating Euclidean distance:

```text
age difference     =     5  → small contribution
charges difference = 10,000 → dominates the result
```

---

**StandardScaler**

`StandardScaler` transforms each feature so that:

```text
mean = 0
standard deviation = 1
```

For example, original age values such as 20, 30, 40, 50
might become -1.3, -0.4, 0.5, 1.4 after scaling.
The exact values depend on the mean and standard deviation of the feature.

After scaling:

```text
age      → comparable scale
bmi      → comparable scale
children → comparable scale
charges  → comparable scale
```

All features now contribute equally to distance calculations.

This is the same `StandardScaler` discipline from Week 4 —
the reason changes (distance vs. penalty) but the requirement is identical.

## 6. Choosing k — The Elbow Method

K-Means requires choosing the number of clusters, `k`, in advance.
But the correct value of `k` is usually not known beforehand.

The **Elbow Method** gives an empirical way to choose a reasonable value of `k`.

---

**How it works:**

Run K-Means for k = 1 to 10. Record the inertia at each k.
Then plot inertia vs k and look for the **elbow** — the point
where adding more clusters gives diminishing returns.

```text
Inertia
│
│●
│  ●
│    ●
│      ●
│        ●        ← elbow
│          ●
│            ● ● ●
│
└────────────────────── k
  1  2  3  4  5  6  7
```

---

**How to interpret the elbow:**

Before the elbow: adding another cluster produces a large reduction
in inertia — each additional cluster captures meaningful structure.

After the elbow: adding another cluster produces only a small
reduction — diminishing returns.

The elbow point is therefore a reasonable candidate for k.

---

**An important limitation:**

The elbow is not always obvious. Sometimes the curve does not
contain a clear bend, and several values of k may look reasonable.

That is why I use the Silhouette Score as a second quantitative
check rather than relying only on the elbow.


---

## 7. Choosing k — The Silhouette Score

The **Silhouette Score** is a quantitative measure of how well
data points fit within their assigned clusters.

It answers the question:

> **"Is each point closer to its own cluster than to the nearest other cluster?"**

---

**How it is calculated:**

For each data point, two quantities are computed:

- **a** = average distance to all other points in the same cluster
- **b** = average distance to all points in the nearest other cluster

```text
silhouette = (b - a) / max(a, b)
```

For a well-clustered point, **a** should be small and **b** should be large — meaning the point
is close to its own cluster and far from the nearest competing cluster.

---

**Understanding the score:**

```text
+1  → point is well inside its cluster and far from others
 0  → point is on the boundary between two clusters
-1  → point is closer to another cluster than its own
```

**A worked example:**

Suppose a point has a = 2 and b = 10:

```text
silhouette = (10 - 2) / max(2, 10)
           = 8 / 10
           = 0.8
```

A score of 0.8 means this point is well matched to its cluster.

---

**Interpreting the mean silhouette score:**

| Score range | Meaning |
|------------|--------|
| 0.70 – 1.00 | Strong structure |
| 0.51 – 0.69 | Reasonable structure |
| 0.26 – 0.50 | Weak structure — clusters overlap |
| < 0.25 | No meaningful structure |

These are general guidelines, not universal rules.
The meaning of a score depends on the dataset and the problem.

---

**Using elbow and silhouette together:**

```text
Elbow Method
     ↓
Candidate values of k
     ↓
Silhouette Score
     ↓
Compare candidates quantitatively
     ↓
Choose k
```

The highest silhouette score is useful evidence, but it should
not be treated as the only decision rule. Visualization and the
real-world meaning of the clusters should also be considered.

## 8. Interpreting Clusters

K-Means assigns a cluster label to every data point.
But a label like `0`, `1`, or `2` has no meaning on its own.

The final step — and the most important one — is to understand
**what each cluster actually represents** in the real world.

---

**How interpretation works:**

After fitting K-Means, I add the cluster labels back to the
original (unscaled) dataframe and use `groupby` to compute
the mean of each feature per cluster:

```text
Original data + cluster labels
        ↓
groupby(cluster label).mean()
        ↓
Profile of each cluster
```

For example, on the insurance dataset:

```text
Cluster | age  | bmi  | charges | smoker_rate
--------+------+------+---------+------------
   0    | 28.4 | 26.1 |  8,200  |    0.04
   1    | 44.7 | 33.8 | 41,500  |    0.82
   2    | 41.2 | 29.3 | 11,900  |    0.08
```

From this table, even without knowing the cluster labels in advance,
it's possible to assign meaning to each group:

- **Cluster 0** → young, healthy, low cost
- **Cluster 1** → high BMI, high smoker rate, high cost
- **Cluster 2** → middle-aged, moderate BMI, non-smoker, moderate cost

---

**Why this step is not automatic:**

The algorithm assigns numbers — not names.
The meaning of each cluster comes from reading the feature profiles
and applying domain knowledge.

Two different runs with different k values or different datasets
will produce completely different clusters.
There is no universal "Cluster 1 = low risk" label.

---

**A complete K-Means workflow:**

```text
Load & clean data
       ↓
Scale features (StandardScaler)
       ↓
Elbow method → candidate k values
       ↓
Silhouette score → confirm best k
       ↓
Fit KMeans(k=best_k, n_init=10)
       ↓
Add cluster labels to original dataframe
       ↓
groupby(cluster).mean() → interpret each segment
```
