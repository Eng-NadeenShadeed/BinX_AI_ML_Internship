### Import NumPy

```python
import numpy as np
```

### Step 1: 
Represent three data samples as a (3 × features) NumPy matrix.

```python
# Three samples
X = np.array([
    [25, 50000, 3],
    [40, 80000, 10],
    [33, 62000, 5]
])

print(X)

print("Shape:", X.shape)
```

### Step 2: 
Compute the dot product of one sample vector with a weight vector by hand, then verify it with
np.dot.


**Create Weight Vector**

```python
# Model weights
weights = np.array([
    0.1,
    0.0002,
    1.5
])

print(weights)

print("Shape:", weights.shape)
```

**Dot Product Manually**

```python
sample = X[0]

manual_prediction = (
    sample[0] * weights[0]
    +
    sample[1] * weights[1]
    +
    sample[2] * weights[2]
)

manual_prediction
```

**Verify using NumPy**

```python
numpy_prediction = np.dot(sample, weights)

numpy_prediction
```

### Step 3: 
Use matrix multiplication (@) to produce a prediction for all three samples at once.

```python
predictions = X @ weights

predictions
```

### Step 4: 
Deliberately create a shape-mismatch error, read the message, and explain in Markdown why it
occurred and how to fix it.


```python
wrong_weights = np.array([
    0.1,
    0.0002
])

X @ wrong_weights
```

**Understanding Shape Errors**

The matrix:

X

has shape:

(3,3)

meaning:

3 samples
3 features


The weights vector must contain:

3 values

because every feature needs one weight.


Using only 2 weights creates a mismatch:

(3,3) @ (2,)


The inner dimensions do not match.

Correct:

(3,3) @ (3,)

**Fix the Error**

```python
correct_weights = np.array([
    0.1,
    0.0002,
    1.5
])


X @ correct_weights
```
