### Probability

Probability is a branch of mathematics that measures **how likely an event is to happen**.

It helps us deal with **uncertainty**, which exists in many real-world problems.

In Machine Learning, probability is essential because data is rarely perfect or completely predictable. Models often estimate the probability of an event instead of giving a definite answer.

For example:

- Email spam detection
- Disease diagnosis
- Fraud detection
- Weather prediction

All of these problems involve uncertainty, so probability helps machines make better decisions.

### Probability Scale

A probability is always a value between **0 and 1**.

- **0** → Impossible event.
- **1** → Certain event.
- **Between 0 and 1** → The event may happen.

Examples:

- Probability that the sun rises tomorrow = **1**
- Probability of rolling a 7 on a six-sided die = **0**
- Probability of getting Heads from a fair coin = **0.5**

### Probability Formula

The probability of an event is calculated as:

$$
P(A)=\frac{\text{Number of Favorable Outcomes}}{\text{Total Number of Possible Outcomes}}
$$

Where:

- **P(A)** means the probability of event A.
- Favorable outcomes are the outcomes we want.
- Total outcomes are all possible outcomes.

### Example

A fair coin has two possible outcomes:

- Heads
- Tails

If we want the probability of getting Heads:

Favorable outcomes = 1

Total outcomes = 2

Therefore:

$$
P(\text{Heads})=\frac{1}{2}=0.5
$$

There is a **50% chance** of getting Heads.

## Basics of Probability

Probability is built on four important concepts.

### 1. Sample Space (S)

The **Sample Space** is the set of **all possible outcomes** of an experiment.

It is represented by the symbol **S**.

Example:

Rolling one six-sided die:

$$
S=\{1,2,3,4,5,6\}
$$

Since there are six possible outcomes, the sample space contains six elements.

### 2. Event (E)

An **Event** is one or more outcomes selected from the sample space.

In other words:

An event is a subset of the sample space.

Example:

Rolling an even number.

Possible outcomes are:

$$
E=\{2,4,6\}
$$

This event contains three possible outcomes.

### 3. Random Experiment

A **Random Experiment** is an experiment whose result is unknown before performing it.

Examples:

- Tossing a coin
- Rolling a die
- Drawing a card
- Selecting a random student

The outcome changes every time the experiment is repeated.

### 4. Random Variable (RV)

A **Random Variable** assigns a numerical value to every possible outcome of an experiment.

There are two types of random variables.



#### Discrete Random Variable

A discrete random variable can take **countable values**.

Examples:

- Number of heads in 10 coin tosses.
- Number of students in a classroom.

Possible values:

## Probability Rules

### 1. Addition Rule (Union)

The Addition Rule calculates the probability that **either event A or event B happens**.

Formula:

$$
P(A\cup B)=P(A)+P(B)-P(A\cap B)
$$

Why do we subtract the intersection?

Because if an outcome belongs to both events, it would be counted twice.

### Example

Suppose we roll a fair six sided die once.

Let:

- **A** = Rolling an even number

$$
A = \{2,4,6\}
$$

- **B** = Rolling a number greater than 3

$$
B = \{4,5,6\}
$$

The intersection of the two events is:

$$
A \cap B = \{4,6\}
$$

The union of the two events is:

$$
A \cup B = \{2,4,5,6\}
$$

Notice that **4** and **6** belong to both events. When calculating the probability of **A or B**, these outcomes would be counted twice if we simply added \(P(A)\) and \(P(B)\). Therefore, we subtract the intersection once to avoid double counting.

The Addition Rule is:

$$
P(A \cup B) = P(A) + P(B) - P(A \cap B)
$$

````python
# Verifying the die roll example above with code
P_A = 3/6   # A = even numbers {2,4,6}
P_B = 3/6   # B = numbers greater than 3 {4,5,6}
P_A_and_B = 2/6   # intersection {4,6}

P_A_or_B = P_A + P_B - P_A_and_B
print("P(A or B):", P_A_or_B)
````

### 2. Multiplication Rule

This rule calculates the probability that **both events happen together**.

For independent events:

$$
P(A\cap B)=P(A)\times P(B)
$$

Independent means:

The occurrence of one event does **not affect** the other.

### Example

Suppose we flip a fair coin and then roll a fair six-sided die.

The probability of getting **Heads** is:

$$
P(\text{Heads})=\frac{1}{2}
$$

The probability of rolling a **6** is:

$$
P(\text{Rolling }6)=\frac{1}{6}
$$

Since the two events are **independent**, we use the Multiplication Rule:

$$
P(\text{Heads} \cap \text{Rolling }6)=P(\text{Heads}) \times P(\text{Rolling }6)
$$

Substitute the values:

$$
P(\text{Heads} \cap \text{Rolling }6)=\frac{1}{2}\times\frac{1}{6}
$$

Therefore:

$$
P(\text{Heads} \cap \text{Rolling }6)=\frac{1}{12}
$$

This means that the probability of getting **Heads** on the coin toss **and** rolling a **6** on the die is **1/12**.

````python
# Verifying the coin + die example above with code
P_heads = 1/2
P_six = 1/6

P_heads_and_six = P_heads * P_six
print("P(Heads and 6):", P_heads_and_six)
````

### 3. Complement Rule

Sometimes it is easier to calculate the probability that an event **does not happen**.

Formula:

$$
P(A')=1-P(A)
$$

Where:

A' means "NOT A".

### Example

Suppose the probability of rain is:

$$
P(\text{Rain}) = 0.3
$$

The probability of **no rain** is the complement of the event **Rain**.

Using the Complement Rule:

$$
P(\text{No Rain}) = 1 - P(\text{Rain})
$$

Substitute the value:

$$
P(\text{No Rain}) = 1 - 0.3
$$

Therefore:

$$
P(\text{No Rain}) = 0.7
$$

This means there is a **70% probability** that it will **not rain**.

````python
# Verifying the rain example above with code
P_rain = 0.3
P_no_rain = 1 - P_rain
print("P(No Rain):", P_no_rain)
````

### 4. Independent Events

Two events are independent if one event does not affect the other.

Examples:

- Tossing a coin and rolling a die.
- Choosing a card then rolling a die.

For independent events:

$$
P(A\cap B)=P(A)\times P(B)
$$

### 5. Conditional Probability

Conditional Probability means:

"What is the probability of A **given that** B has already happened?"

Formula:

$$
P(A|B)=\frac{P(A\cap B)}{P(B)}
$$

The symbol "|" means **given that**.

### Example

Suppose a class has:

- **20 students**
- **8 females**
- Among the females, **5 study Engineering**

If we already know that the selected student is **female**, what is the probability that she studies **Engineering**?

This is a **conditional probability** because we are calculating the probability of one event given that another event has already occurred.



Let:

- **A** = The student studies Engineering.
- **B** = The student is Female.

Since we already know the student is female, our sample space is no longer all **20 students**. Instead, it is limited to the **8 female students**.

Therefore:

$$
P(\text{Engineering} \mid \text{Female})=\frac{5}{8}
$$

**Not:**

$$
\frac{5}{20}
$$

because we are only considering the group of **female students**, not the entire class.

This means that the probability of a student studying **Engineering**, given that the student is **female**, is:

$$
P(\text{Engineering} \mid \text{Female})=\frac{5}{8}=0.625
$$

````python
# Verifying the class example above with code
total_students = 20
female_students = 8
female_engineering = 5

P_engineering_given_female = female_engineering / female_students
print("P(Engineering | Female):", P_engineering_given_female)
````

### 6. Bayes' Theorem

Bayes' Theorem updates the probability of an event after receiving new information.

Formula:

$$
P(A|B)=\frac{P(B|A)\times P(A)}{P(B)}
$$

It combines:

- Prior Probability
- New Evidence
- Updated Probability (Posterior)

#### The Main Idea Behind Bayes' Theorem

The idea behind **Bayes' Theorem** is simple:

1. Start with an initial belief or probability, called the **Prior**.
2. Receive new information or evidence, called the **Evidence**.
3. Update your belief using **Bayes' Theorem** to obtain a more accurate probability, called the **Posterior**.


In other words, Bayes' Theorem allows us to **update probabilities whenever new evidence becomes available**.


#### Why is Bayes' Theorem Important in Machine Learning?

Bayes' Theorem is widely used in Machine Learning because models often need to **continuously update their predictions** as new data becomes available.

Some common applications include:

-  **Spam Email Detection**
-  **Medical Diagnosis**
-  **Fraud Detection**
-  **Sentiment Analysis**
-  **Naive Bayes Classifier**

Bayesian methods enable machine learning models to improve their predictions by combining **prior knowledge** with **new evidence**, making them especially useful in situations involving uncertainty.

### Worked Example: A Medical Test

Suppose a disease affects **1%** of the population. A test for it is 
**99% accurate** for people who have the disease (true positive rate), 
but has a **5% false positive rate** for people who don't have it.

If someone tests positive, what's the actual probability they have the 
disease? Intuition says "close to 99%" - but Bayes' theorem tells a very 
different story.

````python
P_disease = 0.01
P_positive_given_disease = 0.99
P_positive_given_no_disease = 0.05
P_no_disease = 1 - P_disease

# P(Positive) - total probability of testing positive, from both groups
P_positive = (P_positive_given_disease * P_disease) + (P_positive_given_no_disease * P_no_disease)

# Bayes' Theorem: P(Disease | Positive)
P_disease_given_positive = (P_positive_given_disease * P_disease) / P_positive

print("P(Positive):", P_positive)
print("P(Disease | Positive):", P_disease_given_positive)
print("As percentage:", round(P_disease_given_positive * 100, 2), "%")
````

Even with a 99% accurate test, a positive result only means about a 
**16.67% chance** of actually having the disease - not 99%. This is 
because the disease is rare (1%), so most positive results actually come 
from the much larger group of healthy people who got a false positive. 
This is exactly why the prior probability P(Disease) matters so much in 
Bayes' theorem - ignoring it leads to a very wrong intuition.

## Normal Distribution

Normal Distribution is a continuous probability distribution that is symmetric around its mean.

It is also called:

- Gaussian Distribution
- Bell Curve

In a normal distribution:

- Values close to the mean occur more frequently.
- Values far from the mean occur less frequently.

The graph has a bell shape.

### Example of Normal Distribution

Imagine a class of students taking a math exam.

Most students get scores close to the average.

Few students get very low scores.

Few students get very high scores.

The distribution of scores will look like a bell-shaped curve.

Example:

Mean score = 75

Most students:

70 - 80

Few students:

40 or 100

### Probability Density Function (PDF)

The formula of Normal Distribution:

$$
f(x)=
\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$


Where:

x = value of the random variable

μ = mean of the distribution

σ = standard deviation

π ≈ 3.14159

e ≈ 2.71828

### Characteristics of Normal Distribution

### 1. Symmetry

The normal distribution is symmetric around the mean.

The left side is a mirror image of the right side.


### 2. Mean, Median, and Mode

In a normal distribution:

Mean = Median = Mode

All are located at the center.


### 3. Bell-Shaped Curve

Most observations are concentrated around the center.

Values become less frequent as they move away from the mean.


### 4. Asymptotic Tails

The curve never touches the x-axis.

It extends infinitely in both directions.

### Parameters of Normal Distribution

A normal distribution is determined by two parameters:

#### 1. Mean (μ)

The mean controls the center of the distribution.

Increasing μ:

The curve moves to the right.

Decreasing μ:

The curve moves to the left.


#### 2. Standard Deviation (σ)

The standard deviation controls the spread.

Small σ:

- Narrow curve
- Data is close to the mean


Large σ:

- Wider curve
- Data is more spread out

### Empirical Rule (68-95-99.7 Rule)

The Empirical Rule explains how data is distributed around the mean.

#### Within 1 Standard Deviation

Approximately:

68%

of data lies between:

$$
\mu-\sigma
$$

and

$$
\mu+\sigma
$$


#### Within 2 Standard Deviations

Approximately:

95%

of data lies between:

$$
\mu-2\sigma
$$

and

$$
\mu+2\sigma
$$


#### Within 3 Standard Deviations

Approximately:

99.7%

of data lies between:

$$
\mu-3\sigma
$$

and

$$
\mu+3\sigma
$$

### Applications of Normal Distribution in Machine Learning

#### 1. Feature Scaling

Z-score normalization uses:

$$
Z=\frac{x-\mu}{\sigma}
$$

It transforms features to have:

Mean = 0

Standard deviation = 1


#### 2. Anomaly Detection

Values far from the mean may represent unusual behavior.


#### 3. Bayesian Models

Gaussian distributions are used as prior distributions.


#### 4. Gaussian Mixture Models

Used in clustering and unsupervised learning.


#### 5. Image Processing

Gaussian filters are used to reduce image noise.


#### 6. Generative AI

Diffusion models use Gaussian noise during image generation.

### Example: Detecting Anomalies

Suppose we have student exam scores:

Mean:

μ = 70

Standard deviation:

σ = 10


Most students should have scores between:

60 and 80


A student with:

Score = 110


is far from the mean.

The model may consider this value as an anomaly.

### Z-Score

Z-score tells us how many standard deviations a value is away from the mean.

Formula:

$$
Z=\frac{x-\mu}{\sigma}
$$


Where:

x = value

μ = mean

σ = standard deviation

### Example

Mean = 100

Standard deviation = 15

Value = 130


$$
Z=\frac{130-100}{15}
$$

$$
Z=2
$$


Interpretation:

The value 130 is 2 standard deviations above the mean.

##  Binomial Distribution

### Idea

Binomial Distribution is a **discrete probability distribution** that models:

> The number of successes in a fixed number of independent binary trials.

In simple words:

We want to know:

**How many times does success happen during a fixed number of attempts?**

---

### Example: Coin Flips

We flip a coin 10 times.

Each trial has two possible outcomes:


Head → Success

Tail → Failure


Question:

What is the probability of getting exactly 5 Heads?

This is a Binomial Distribution problem.

---

### Conditions of Binomial Distribution

To use the **Binomial Distribution**, four conditions must be satisfied.


#### 1) Fixed Number of Trials

The number of experiments or trials must be fixed.

Example:

Flipping a coin 10 times.

$$
n = 10
$$

The number of trials is known before starting the experiment.

Incorrect example:

Flipping a coin until getting 5 heads.

Because the number of trials is not fixed.

---

#### 2) Two Possible Outcomes

Each trial must have only two possible outcomes.

These outcomes are usually called:

- Success
- Failure

Example:

Coin flip:

- Head = Success
- Tail = Failure

Other examples:

- Email classification:
  - Spam = Success
  - Not Spam = Failure

- Machine learning prediction:
  - Correct prediction = Success
  - Wrong prediction = Failure

---

#### 3) Independent Trials

Each trial must be independent.

This means:

The result of one trial does not affect the result of another trial.

Example:

When flipping a coin:

The first flip does not change the probability of the second flip.

Each flip remains:

$$
P(Head)=0.5
$$

---

#### 4) Constant Probability

The probability of success must remain the same for every trial.

Example:

For a fair coin:

$$
P(Head)=0.5
$$

The probability of getting Head is constant in every flip.

Therefore:

$$
P(Failure)=1-p
$$

For a fair coin:

$$
P(Tail)=1-0.5=0.5
$$


---

### Binomial Formula Intuition

The Binomial Distribution formula is:

$$
P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}
$$


Where:

- $n$ = number of trials
- $k$ = number of successes
- $p$ = probability of success
- $(1-p)$ = probability of failure

---

### Parameters Explanation

##### n: Number of Trials

$n$ represents the total number of experiments.

Example:

Flipping a coin 10 times:

$$
n=10
$$


---

##### k: Number of Required Successes

$k$ represents how many successes we want.

Example:

We want exactly 5 Heads:

$$
k=5
$$


---

##### p: Probability of Success

$p$ represents the probability of success in one trial.

For a fair coin:

$$
p=0.5
$$


---

##### (1-p): Probability of Failure

The probability of failure is:

$$
1-p
$$

For a fair coin:

$$
1-0.5=0.5
$$


---

### Manual Example

#### Problem

A coin is flipped 10 times.

The probability of getting Head is:

$$
p=0.5
$$

Find the probability of getting exactly 5 Heads.

---

##### Given Parameters

### Example 1: Probability of Exactly 5 Heads

A fair coin is flipped 10 times.

Find the probability of getting exactly 5 Heads.

Given:

Number of trials:

$$
n=10
$$

Probability of success:

$$
p=0.5
$$

Number of required successes:

````python
from scipy.stats import binom

n = 10
p = 0.5
k = 5

prob = binom.pmf(k, n, p)

print(prob)
````

**What is PMF?**

PMF stands for:

**Probability Mass Function**

It gives the probability of a specific value in a discrete probability distribution.

Example:

Finding the probability of getting exactly:

- 5 Heads
- 3 successes
- 10 correct predictions

PMF answers questions like:

$$
P(X=k)
$$

where we want one specific value.

### Example 2: Probability of At Least 5 Heads

At least 5 Heads means:

$$
X \geq 5
$$

Possible values:

5, 6, 7, 8, 9, 10

To calculate this, we use:

`sf()`

which means:

Survival Function

````python
prob = binom.sf(4, n, p)

print(prob)
````

**Why do we use 4?**

The survival function is:

$$
sf(x)=P(X>x)
$$


We need:

$$
P(X \geq 5)
$$


So we rewrite it as:

$$
P(X>4)
$$


Therefore:

$$
X>4
$$

includes:

5, 6, 7, 8, 9, 10



**sf() calculates everything after a certain value.**

Instead of adding manually:

P(5)+P(6)+P(7)+...+P(10)

we simply use:

binom.sf(4,n,p)

## Uniform Distribution

A Uniform Distribution is a probability distribution where all possible outcomes have the same probability.

In other words:

> Every value has an equal chance of occurring.

The probability is distributed evenly across the possible outcomes.

Uniform Distribution can be:

1. Discrete Uniform Distribution
2. Continuous Uniform Distribution

The main difference depends on whether the possible values are countable or can take any value inside an interval.

### Why Does Uniform Distribution Have Equal Probability?

The idea behind Uniform Distribution is fairness.

No value is preferred over another value.

For example:

A fair dice:

P(1) = P(2) = P(3) = P(4) = P(5) = P(6)

All outcomes are equally likely.

For a continuous uniform distribution:

The probability density remains constant over the entire range.

The distribution does not have a peak like the Normal Distribution.

### Continuous Uniform Distribution Formula

For a continuous uniform distribution:

X ~ Uniform(a, b)

where:

a = minimum value

b = maximum value


The probability density function (PDF):

f(x) = 1 / (b - a)

for:

a <= x <= b


The total area under the distribution curve is always equal to 1.


#### Formula Intuition

The height of the distribution depends only on the interval length.

If the interval is large:

**Example:**

0 to 100

The probability density becomes smaller.

Because the same total probability (1) is spread over a larger area.


If the interval is small:

**Example:**

0 to 10

The density becomes larger.

Because probability is concentrated in a smaller range.

### Discrete Uniform Distribution Formula

For n possible outcomes:

P(X = x) = 1 / n


**Example:**

- Rolling a dice:

    Number of outcomes:

    n = 6


- Probability:

    P(X = 3)

    = 1 / 6

    = 0.1667

### Uniform Distribution Using Python

We use:

**scipy.stats.uniform**

because SciPy provides statistical distributions and functions.

Import Libraries

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform
````

#### Generate Random Samples

We can generate random values from a uniform distribution using:

uniform.rvs()

Parameters:

- loc = starting point

- scale = range length

````python
samples = uniform.rvs(
    loc=0,
    scale=10,
    size=1000,
    random_state=42
)

samples[:10]
````

### Visualizing Uniform Distribution

````python
plt.figure(figsize=(8,5))

plt.hist(
    samples,
    bins=30,
    density=True
)

plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Probability Density")

plt.show()
````

The histogram shows that:

- Values are spread evenly.
- There is no central peak.
- Each region has approximately the same probability.

This is different from Normal Distribution where values near the mean occur more frequently.

### Probability Density Function (PDF)

The PDF gives the density value at a specific point.

For continuous distributions:

we calculate density, not probability of one exact value.

````python
density = uniform.pdf(
    x=5,
    loc=0,
    scale=10
)

density
````

For continuous uniform distribution:

Probability between two points:

P(a < X < b)

can be calculated using:

CDF(b) - CDF(a)

````python
probability = uniform.cdf(
    8,
    loc=0,
    scale=10
) - uniform.cdf(
    3,
    loc=0,
    scale=10
)

probability
````

### Uniform Distribution in ML Applications

Uniform Distribution appears frequently in Machine Learning.

Main applications:

1. Random Initialization
2. Random Sampling
3. Hyperparameter Search

### 1. Random Initialization

Neural networks start with random weights.

A common strategy is:

Initialize weights from a uniform distribution.

**Example:**

Weights are sampled between:

-0.5 and 0.5


Why?

Because starting with identical weights causes neurons to learn the same features.

Random initialization helps break symmetry.

````python
weights = np.random.uniform(
    low=-0.5,
    high=0.5,
    size=(3,3)
)

weights
````

### 2. Random Sampling

Uniform distribution is used when we want random samples where every value has equal chance.

Examples:

- Train/test data sampling
- Simulation
- Data augmentation
- Monte Carlo methods

````python
random_numbers = np.random.uniform(
    0,
    1,
    10
)

random_numbers
````

### 3. Hyperparameter Optimization

When searching for the best ML model parameters:

**Examples:**

- Learning rate:

     0.0001 --> 0.1


- Dropout:

     0.1 --> 0.5


We can randomly sample values from a uniform distribution.

This is used in Random Search algorithms.

## Uniform vs Normal Distribution

Uniform Distribution:

- All values have equal probability.
- Flat distribution shape.
- No preferred value.


Normal Distribution:

- Values near the mean are more common.
- Bell-shaped curve.
- Extreme values are rare.

------------------------------------------------------------------------------------------------------------------------------

## Normal vs Binomial vs Uniform Distribution in ML


| Distribution | Type | Main Idea | Example | ML Usage |
|--------------|------|-----------|---------|----------|
| Normal Distribution | Continuous | Values cluster around mean | Heights, measurement errors | Data modeling, noise modeling, feature analysis |
| Binomial Distribution | Discrete | Number of successes in fixed trials | Coin flips, classification outcomes | Classification probabilities, binary events |
| Uniform Distribution | Continuous/Discrete | Every outcome has equal probability | Dice roll, random numbers | Random initialization, sampling, hyperparameter search |


### Key Differences:

Normal:
- Has mean and standard deviation.
- Bell-shaped curve.
- Most values are near the center.


Binomial:
- Counts successes.
- Requires fixed number of trials.
- Only two possible outcomes.


Uniform:
- Every value is equally likely.
- Flat distribution.
- Used for random generation.
