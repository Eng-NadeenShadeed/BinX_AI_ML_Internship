## Exploratory Data Analysis (EDA)

### What is Exploratory Data Analysis?

**Exploratory Data Analysis (EDA)** is the process of exploring, summarizing, and visualizing a dataset before building any Machine Learning or statistical model.

The purpose of EDA is to understand the data, discover hidden patterns, identify potential problems, detect outliers, and examine relationships between variables.

EDA is one of the most important stages of any Data Science or Machine Learning project because the quality of the model depends heavily on understanding the data first.

---

### Why is EDA Important?

EDA helps us answer important questions about the dataset before training a model.

It allows us to:

- Understand the dataset structure.
- Identify numerical and categorical features.
- Check data types.
- Detect missing values.
- Detect duplicate records.
- Discover outliers.
- Understand data distributions.
- Find relationships between variables.
- Select useful features.
- Decide which preprocessing techniques are needed.
- Choose appropriate Machine Learning algorithms.

> **Golden Rule in Data Science:**
>
> **Never build a Machine Learning model before performing EDA.**

---

### Why Does Machine Learning Need EDA?

Machine Learning algorithms do not understand the meaning of data.

They only learn from numbers.

If the dataset contains:

- Missing values
- Wrong data types
- Duplicate rows
- Extreme outliers
- Highly skewed distributions

the model will learn incorrect patterns and produce poor predictions.

EDA helps us detect and fix these issues before training the model.

---

### Main Goals of EDA

The main objectives of Exploratory Data Analysis are:

1. Understand the dataset.
2. Discover hidden patterns.
3. Detect anomalies and outliers.
4. Find relationships between variables.
5. Validate assumptions.
6. Prepare the data for Machine Learning.


---

### Types of Exploratory Data Analysis (EDA)

EDA can be categorized into three main types depending on the number of variables being analyzed.

---

#### 1. Univariate Analysis

Univariate Analysis focuses on analyzing **one variable at a time**.

The goal is to understand the distribution, central tendency, spread, and possible outliers of a single feature.

Common questions answered:

- How are the values distributed?
- Is the data symmetric or skewed?
- Are there any outliers?
- What is the most common value?
- What is the spread of the data?

Common visualizations:

- Histogram
- KDE Plot
- Box Plot
- Count Plot (for categorical variables)

> **Today's lesson focuses on Univariate Analysis because it is the first step of Exploratory Data Analysis.**

---

#### 2. Bivariate Analysis

Bivariate Analysis studies the relationship between **two variables**.

Its objective is to determine whether two variables are related and how strongly they influence each other.

Examples:

- Scatter Plot
- Correlation Analysis
- Line Plot
- Cross Tabulation


---

#### 3. Multivariate Analysis

Multivariate Analysis explores relationships among **three or more variables simultaneously**.

It helps discover complex interactions that cannot be observed when studying variables individually.

Common techniques include:

- Pair Plot
- Heatmap
- PCA (Principal Component Analysis)

Advanced Machine Learning projects often rely on multivariate analysis to better understand the dataset.

---

#### Summary

| Type | Number of Variables | Main Goal |
|------|----------------------|-----------|
| Univariate | 1 | Understand one feature |
| Bivariate | 2 | Study relationships |
| Multivariate | 3+ | Analyze complex interactions |

## Histogram

## What is a Histogram?

A **Histogram** is a visualization used to show the **distribution of a numerical variable**.

Instead of displaying every individual value, a histogram groups the data into intervals called **bins** and counts how many observations fall inside each interval.

This allows us to quickly understand how the data is distributed.

---

### Why Do We Use a Histogram?

A histogram helps answer several important questions about a dataset:

- How are the values distributed?
- Where are most observations concentrated?
- Is the distribution symmetric or skewed?
- Are there multiple peaks (multimodal)?
- Are there unusually large or small values?
- Does the data resemble a normal distribution?

Instead of reading thousands of numbers, we can understand the overall shape of the data with a single chart.

---

### How Does a Histogram Work?

Suppose we have the following exam scores:

70, 72, 75, 77, 78, 80, 81, 83, 84, 85, 87, 90, 92, 95

Instead of plotting every value individually, the histogram groups them into ranges.

| Score Range | Frequency |
|-------------|-----------|
| 70–75 | 3 |
| 76–80 | 3 |
| 81–85 | 5 |
| 86–90 | 2 |
| 91–95 | 2 |

Each bar represents a range of values, and the height of the bar represents how many observations fall within that range.

---

### Understanding the Axes

**X-axis**
- Represents the numerical values divided into intervals (bins).

**Y-axis**
- Represents the frequency (count) of observations inside each interval.

---

### What Can We Learn From a Histogram?

A histogram helps us identify:

#### 1. Center

Where most values are concentrated.

Example:
Most customer ages are between **25 and 35 years**.

---

#### 2. Spread

How widely the data is distributed.

- Narrow histogram → low variability.
- Wide histogram → high variability.

---

#### 3. Skewness

Whether the distribution is balanced or biased toward one side.

- Symmetric
- Right-skewed (Positive Skew)
- Left-skewed (Negative Skew)

---

#### 4. Multiple Peaks

Sometimes the histogram has two or more peaks.

This may indicate that the dataset contains different groups or populations.

---

#### 5. Possible Outliers

Bars that appear far away from the majority of the data may indicate unusual observations that require further investigation.

A histogram suggests possible outliers, but a **Box Plot** is a better tool for identifying them.

---

### Histogram in Machine Learning

Before training a Machine Learning model, we often examine the distribution of every numerical feature.

This helps us decide whether we need to:

- Normalize the data.
- Standardize the data.
- Apply logarithmic transformations.
- Handle skewed distributions.
- Investigate unusual values.

Understanding feature distributions leads to better preprocessing decisions and often improves model performance.

> A histogram is usually the **first visualization** created during Univariate Exploratory Data Analysis.

---

**Example using Seaborn**

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load example dataset
tips = sns.load_dataset("tips")

# Create histogram
sns.histplot(data=tips, x="total_bill")

plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")

plt.show()
```

**Dataset Description**

The **tips** dataset is a built in dataset provided by Seaborn.
It contains restaurant billing information including total bill,
tip amount, customer information, and visit details.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.histplot(
    data=tips,
    x="total_bill",
    kde=True #KDE Curve
)

plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")

plt.show()
```

**Interpretation**

From the histogram we can observe:

- Most customers have relatively small or medium bills.
- Very large bills occur less frequently.
- The distribution is not perfectly symmetric.
- The dataset appears to be positively (right) skewed.
- The KDE curve provides a smooth representation of the overall distribution, making the shape easier to interpret.

## KDE Plot (Kernel Density Estimation)

### What is KDE Plot?

A **KDE (Kernel Density Estimation) Plot** is a visualization technique used to estimate the **probability distribution of a numerical variable**.

Instead of grouping values into bins like a histogram, KDE creates a smooth curve that represents where data points are concentrated.

In simple words:

- Histogram → Counts observations inside intervals.
- KDE Plot → Estimates the shape of the distribution as a smooth curve.

---

### Why Do We Use KDE?

KDE helps us understand the underlying distribution of the data.

It allows us to observe:

- Where most values are concentrated.
- The shape of the distribution.
- Whether the data is symmetric or skewed.
- Whether there are multiple groups inside the data.
- Differences between distributions.

---

### How KDE Works (Intuition)

Imagine every data point adds a small smooth "bump" around its location.

All these bumps are combined together to create one continuous curve.

The higher the curve:

→ More observations are concentrated in that region.

The lower the curve:

→ Fewer observations exist in that region.

---

### Histogram vs KDE

| Histogram | KDE Plot |
|-----------|----------|
| Uses bars | Uses a smooth curve |
| Counts observations | Estimates probability density |
| Depends on number of bins | Depends on bandwidth |
| Easier for frequency interpretation | Easier for understanding distribution shape |

Both are often used together during EDA because they provide complementary information.

---

### KDE in Machine Learning

Understanding the distribution of features helps us make preprocessing decisions.

For example:

- A highly skewed feature may require transformation.
- A normally distributed feature may work better with some algorithms.
- Multiple peaks may indicate different groups in the data.

KDE helps us understand these characteristics before training a model.

**Creating KDE Plot using Seaborn**

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load example dataset
tips = sns.load_dataset("tips")

# Create KDE plot
sns.kdeplot(
    data=tips,
    x="total_bill"
)

plt.title("KDE Distribution of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Density")

plt.show()
```

**Histogram + KDE Together**

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.histplot(
    data=tips,
    x="total_bill",
    kde=True
)

plt.title("Histogram with KDE Curve")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")

plt.show()
```

### How to Interpret a KDE Plot?

When analyzing a KDE curve, we look for:

#### 1. Peak (Highest Point)

The highest point represents the area where most observations are concentrated.

Example:

Most customers have bills around a specific range.

---

#### 2. Spread

A wider curve means the values are more spread out.

A narrower curve means values are more concentrated.

---

#### 3. Skewness

The tail direction helps identify skewness:

- Long right tail → Positive skew.
- Long left tail → Negative skew.

---

#### 4. Multiple Peaks

Multiple peaks may indicate different groups inside the dataset.

For example:

A dataset containing two different customer groups may create two peaks.

---

### Important Note

KDE is an estimation, not the actual probability.

The curve is generated from the available data, so the result depends on parameters such as bandwidth.

Therefore, KDE should be interpreted together with other EDA tools like:

- Histogram
- Box Plot
- Statistical summaries

## Box Plot (Box-and-Whisker Plot)

### What is a Box Plot?

A **Box Plot** is a statistical visualization used to summarize the distribution of a numerical variable.

It provides a compact view of:

- Center of the data (Median)
- Spread of the data (Quartiles and IQR)
- Range of normal values
- Potential outliers

In EDA, the Box Plot is one of the most important plots because it connects visualization directly with:

- Percentiles
- Quartiles
- IQR (Interquartile Range)
- Outlier Detection


---

### Why Do We Use Box Plot in EDA?

Before building a Machine Learning model, we need to understand our data.

A Box Plot helps us answer questions like:

- Where is the center of the data?
- How spread out are the values?
- Is the distribution symmetric or skewed?
- Are there unusual values (outliers)?
- Does the dataset contain extreme observations?

Example:

If we analyze house prices:

- Most houses may cost between $100k and $500k.
- A few houses may cost $5M.

Those extremely expensive houses will appear as outliers in the Box Plot.


---

### Outlier Detection Using Interquartile Range (IQR)

#### What are Outliers?

Outliers are observations that are significantly different from the general pattern of the dataset.

They can happen because of:
- Data entry mistakes
- Measurement errors
- Rare but real events

Outliers can negatively affect statistical analysis and machine learning models because they may distort:
- Mean
- Variance
- Model parameters

Therefore, detecting and investigating outliers is an important step in Exploratory Data Analysis (EDA).

> An outlier is a question, not a verdict.  
> We should investigate whether it is a real value or an error before deciding to remove it.

### Components of a Box Plot

A Box Plot consists of five main statistical values:

1. Minimum
2. Q1 (First Quartile)
3. Median (Q2)
4. Q3 (Third Quartile)
5. Maximum


These values are called the **Five Number Summary**.


---

### Five Number Summary

Given sorted data:

10, 20, 30, 40, 50, 60, 70, 80



The Box Plot summarizes it as:

| Statistic | Meaning |
|---|---|
| Minimum | Smallest value |
| Q1 | 25% of data is below this value |
| Median | Middle value (50th percentile) |
| Q3 | 75% of data is below this value |
| Maximum | Largest value |


### Understanding Quartiles

Quartiles divide the dataset into four equal parts.


**- Q1 (First Quartile)**

Q1 represents the 25th percentile.

Meaning:

25% of observations are smaller than or equal to Q1.


**- Q2 (Median)**

Q2 represents the 50th percentile.

It divides the data into two equal halves.


**- Q3 (Third Quartile)**

Q3 represents the 75th percentile.

Meaning:

75% of observations are smaller than or equal to Q3.


---

### Interquartile Range (IQR)

The box in a Box Plot represents the middle 50% of the data.

The size of this box is called:

##### IQR (Interquartile Range)


Formula:

$$
IQR = Q3 - Q1
$$


A small IQR means:

- Data values are close together.
- Low variability.


A large IQR means:

- Data values are more spread out.
- High variability.


---

### How Does Box Plot Detect Outliers?

Box Plot uses the IQR rule to identify unusual values.


Lower Bound:

$$
Q1 - 1.5 \times IQR
$$


Upper Bound:

$$
Q3 + 1.5 \times IQR
$$


Any value:

- Smaller than the Lower Bound
- Larger than the Upper Bound

is considered a potential outlier.


---

### Example

Suppose:

Q1 = 20

Q3 = 50


First:

Calculate IQR:

$$
IQR = 50 - 20 = 30
$$


Lower Bound:

$$
20 - 1.5(30)= -25
$$


Upper Bound:

$$
50 + 1.5(30)=95
$$


Therefore:

Any value:

< -25 or > 95

will be detected as an outlier.


---

### Box Plot Interpretation

When reading a Box Plot:

#### 1. Position of Median

If the median is in the center of the box:

→ Data is approximately symmetric.


If the median is closer to Q1:

→ Data is positively skewed.


If the median is closer to Q3:

→ Data is negatively skewed.


#### 2. Size of the Box

Large box:

→ High variability.


Small box:

→ Values are concentrated.


#### 3. Whiskers

Long whiskers:

→ More spread in the data.


Short whiskers:

→ Less spread.


#### 4. Points Outside Whiskers

Points outside:

→ Potential outliers.


---

### Box Plot vs Histogram vs KDE

| Plot | Shows |
|---|---|
| Histogram | Frequency distribution using bins |
| KDE Plot | Smooth estimation of distribution shape |
| Box Plot | Median, quartiles, spread, outliers |

Together they provide a complete understanding of a numerical variable.

```python
# Import Libraries

import seaborn as sns
import matplotlib.pyplot as plt

# Create Example Dataset

data = [10, 12, 15, 18, 20, 22, 25, 30, 35, 100]

# Create Box Plot using Seaborn

sns.boxplot(x=data)

plt.title("Box Plot Example")
plt.xlabel("Values")

plt.show()
```

### Example Dataset
Assume we have the following dataset:

Number of chapatis eaten during lunch:

6, 2, 1, 5, 4, 3, 50

The value 50 looks unusual compared with the rest of the values, so we will use IQR to confirm whether it is an outlier.

**Step 1: Import Libraries**

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

**Step 2: Sort the Data**
IQR calculation requires the dataset to be ordered from smallest to largest.

```python
data = [6, 2, 3, 4, 5, 1, 50]

sorted_data = np.sort(data)

sorted_data
```

**Step 3: Calculate Quartiles and IQR**
We calculate:

- Q1 → 25th percentile
- Q2 → 50th percentile (Median)
- Q3 → 75th percentile

Then:

\[
IQR = Q3 - Q1
\]

```python
data = [6, 2, 3, 4, 5, 1, 50]

Q1 = np.percentile(data, 25, method='midpoint')

Q2 = np.percentile(data, 50, method='midpoint')

Q3 = np.percentile(data, 75, method='midpoint')


print("Q1:", Q1)
print("Q2 (Median):", Q2)
print("Q3:", Q3)


IQR = Q3 - Q1

print("IQR:", IQR)
```

**Step 4: Calculate Outlier Boundaries**
The IQR rule defines two limits:

Lower Bound:

\[
Q1 - 1.5 \times IQR
\]


Upper Bound:

\[
Q3 + 1.5 \times IQR
\]


Any value:

- Smaller than Lower Bound
- Larger than Upper Bound

is considered a potential outlier.

```python
lower_limit = Q1 - 1.5 * IQR

upper_limit = Q3 + 1.5 * IQR


print("Lower Limit:", lower_limit)

print("Upper Limit:", upper_limit)
```

Step 6: Visualize Using Box Plot
A box plot automatically shows:

- Median
- Q1 and Q3
- IQR range
- Potential outliers as points outside the whiskers

```python
sns.boxplot(data=data)

plt.title("Box Plot - Outlier Detection")

plt.show()
```

**Interpretation**

The IQR method detected the value 50 as an outlier because it is above the upper limit (10.0).

However, detecting an outlier does not always mean removing it.

We should investigate:

- Is this value caused by an error?
- Is it a rare but valid observation?

The decision depends on the context of the dataset.

In machine learning projects, outliers are handled carefully.

Possible actions:

1. Keep them:
   - If they represent real-world rare events.

2. Remove them:
   - If they are caused by data errors.

3. Transform or cap them:
   - If they distort the model but contain useful information.

Never remove outliers automatically without understanding the data.

## Count Plot

### What is a Count Plot?

A **Count Plot** is a statistical visualization used to show the frequency (count) of each category in a categorical variable.

It answers the question:

**"How many observations belong to each category?"**

In other words, it counts how many times each category appears in the dataset.

Count Plot is mainly used for:

- Categorical variables
- Class distribution analysis
- Checking imbalance between categories

In EDA, it helps us understand the distribution of categorical features before building ML models.

---

### Why Count Plot Matters?

Before training a machine learning model, we need to understand our data.

For categorical variables, numerical plots like:

- Histogram 
- KDE Plot 
- Box Plot 

are not suitable because they describe numerical distributions.

Instead, Count Plot helps us discover:

- Which categories are most common?
- Are some categories rare?
- Is the dataset balanced or imbalanced?

Example:

If we have a target variable:

| Passed |
|---|
| Yes |
| No |
| Yes |
| Yes |
| No |

A Count Plot will show:

Yes → 3 students  
No → 2 students

---

**Count Plot in Seaborn**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "gender": ["Male", "Female", "Female", "Male", "Male", "Female"],
    "passed": ["Yes", "Yes", "No", "Yes", "No", "Yes"]
}

df = pd.DataFrame(data)

print(df)

sns.countplot(
    data=df,
    x="gender"
)

plt.title("Number of Students by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")

plt.show()
```

---

### Count Plot vs Bar Plot

**Count Plot**

- Counts observations automatically
- Used with categorical variables
- No need to calculate values first

Example:

sns.countplot(data=df, x="gender")


**Bar Plot**

- Shows calculated values
- Requires numerical aggregation

Example:

Average salary by department

---

### Count Plot and Machine Learning

Count plots are very important before modeling because they reveal:

**1. Class imbalance**

Example:

Spam Detection:

Spam:
900 samples

Not Spam:
100 samples


The model may become biased toward the majority class.

**2. Rare categories**

Some categories may have very few examples.

**3. Data quality problems**

Unexpected categories can be detected.
