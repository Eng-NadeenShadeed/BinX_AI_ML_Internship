### Measures of Central Tendency

Measures of Central Tendency are statistical measures used to identify the central or typical value of a dataset. Instead of examining every observation individually, they summarize the dataset with a single representative value.

These measures help answer questions such as:

- What is the average value of the data?
- Which value occurs most frequently?
- What is the middle value after sorting the data?

The three most common measures of central tendency are:

- Mean
- Median
- Mode

Each measure provides a different perspective on the center of the data, and choosing the appropriate one depends on the characteristics of the dataset.

#### Mean (Arithmetic Mean)

The **mean**, commonly known as the average, is the sum of all observations divided by the total number of observations.

It is one of the most widely used statistical measures because it represents the overall center of the dataset.

##### Formula
Mean = ∑x_i/n

Where:

- \(x_i\) = each observation
- \(n\) = total number of observations

The mean works well when the data is approximately symmetric and does not contain extreme outliers. However, it is sensitive to unusually large or small values, which can significantly affect the result.

### Example

Suppose we have the following exam scores:

60, 70, 80, 90, 100

Step 1: Calculate the sum

60 + 70 + 80 + 90 + 100 = 400

Step 2: Divide by the number of observations

Mean = 400 / 5 = 80

Therefore, the average exam score is **80**.

````python
import numpy as np

scores = np.array([60, 70, 80, 90, 100])

mean = np.mean(scores)

print(f"Mean = {mean}")
````

The function `np.mean()` calculates the arithmetic mean by summing all values and dividing by the total number of observations.

In this example:

- Sum of values = 400
- Number of values = 5
- Mean = 80

#### Effect of Outliers

The mean is sensitive to extreme values (outliers).

Consider the following dataset:

60, 70, 80, 90, 500

Although most values are close to each other, the value **500** dramatically increases the average.

````python
scores = np.array([60, 70, 80, 90, 500])

print("Mean =", np.mean(scores))
````

Notice that the average increased from **80** to **160** because of a single extreme value.

This is why the mean is not always the best measure for datasets containing outliers. In such cases, the median often provides a better representation of the center.

### Median

The **median** is the middle value of a dataset after arranging the observations in ascending or descending order.

Unlike the mean, the median is **not affected by extreme values (outliers)**, making it a better measure of central tendency for skewed datasets.

To calculate the median:

- If the number of observations is **odd**, the median is the middle value.
- If the number of observations is **even**, the median is the average of the two middle values.

The median is commonly used in fields such as finance, healthcare, and real estate because these datasets often contain unusually large or small values that can distort the mean.

### Formula

For an **odd** number of observations:

Median = ((n + 1) / 2)th value

For an **even** number of observations:

Median = [(n / 2)th value + ((n / 2) + 1)th value] / 2

> **Important:** The data must always be sorted before finding the median.

### Example 1 (Odd Number of Observations)

Dataset:

15, 10, 25, 20, 30

Step 1: Sort the data

10, 15, 20, 25, 30

There are **5** observations.

The middle value is the **3rd** observation.

Therefore,

**Median = 20**

````python
import numpy as np

numbers = np.array([15, 10, 25, 20, 30])

median = np.median(numbers)

print(f"Median = {median}")
````

### Example 2 (Even Number of Observations)

Dataset:

10, 20, 30, 40

The dataset is already sorted.

There are **4** observations.

The two middle values are:

20 and 30

Median

= (20 + 30) / 2

= 25

Therefore,

**Median = 25**

````python
numbers = np.array([10, 20, 30, 40])

print(np.median(numbers))
````

### Why is the Median Important?

The median is more robust than the mean because it is not influenced by extreme values.

For example, consider the following house prices:

250000, 270000, 280000, 300000, 5000000

The extremely expensive house increases the mean significantly.

However, the median remains close to the typical house price, making it a better representation of the dataset.

````python
prices = np.array([250000, 270000, 280000, 300000, 5000000])

print("Mean  :", np.mean(prices))
print("Median:", np.median(prices))
````

Notice that a single outlier increased the mean dramatically, while the median stayed close to the center of the majority of observations.

This is why the median is often preferred for datasets containing outliers or skewed distributions.

### Mode

The **mode** is the value that appears most frequently in a dataset.

Unlike the mean and median, the mode can be used for both **numerical** and **categorical data** because it focuses on frequency rather than mathematical value.

The mode is especially useful when we want to find the **most common observation** in a dataset.

Examples:

- The most common age group of customers.
- The most frequently purchased product.
- The most common category in a survey.

In statistics:

- A dataset can have **one mode** → Unimodal.
- A dataset can have **two modes** → Bimodal.
- A dataset can have **more than two modes** → Multimodal.
- A dataset with no repeated values has **no mode**.

### Formula

The mode is the value with the highest frequency.

Mode = Most frequently occurring value


There is no mathematical calculation like the mean or median.

Instead, we count how many times each value appears and select the value with the largest count.

### Example 1 (Single Mode)

Dataset:

5, 7, 8, 7, 10, 12, 7

Step 1: Count the frequency of each value.

| Value | Frequency |
|------|-----------|
| 5 | 1 |
| 7 | 3 |
| 8 | 1 |
| 10 | 1 |
| 12 | 1 |

The value **7** appears three times, which is the highest frequency.

Therefore,

**Mode = 7**

````python
import numpy as np
from scipy import stats

numbers = np.array([5, 7, 8, 7, 10, 12, 7])

mode = stats.mode(numbers)

print(mode)
````

### Example 2 (Bimodal Dataset)

A dataset can have more than one mode.

Dataset:

2, 3, 3, 5, 6, 6, 8

Frequency:

| Value | Frequency |
|------|-----------|
| 2 | 1 |
| 3 | 2 |
| 5 | 1 |
| 6 | 2 |
| 8 | 1 |

Both **3** and **6** appear twice.

Therefore,

**Modes = 3 and 6**

This dataset is called:

**Bimodal Dataset**

### Finding Mode in Python

NumPy provides many mathematical functions, but it does not contain a dedicated mode function.

For calculating the mode, we commonly use **SciPy**.

The `scipy.stats.mode()` function returns:

- The mode value.
- The number of occurrences of that value.

````python
from scipy import stats
import numpy as np

data = np.array([2, 3, 3, 5, 6, 6, 8])

result = stats.mode(data)

print("Mode:", result.mode)
print("Frequency:", result.count)
````

### Range

The **Range** is the simplest measure of statistical spread.

It describes the distance between the **maximum value** and the **minimum value** in a dataset.

The range tells us how widely the data values are distributed.

A larger range means the data is more spread out, while a smaller range means the values are closer together.

However, the range is **highly affected by outliers** because it only depends on the smallest and largest values.

### Formula

The formula for the range is:

$$
Range = Maximum\ Value - Minimum\ Value
$$


Where:

- **Maximum Value** → The largest observation in the dataset.
- **Minimum Value** → The smallest observation in the dataset.

### Example

Consider the following dataset:

$$
Data = [10, 12, 15, 18, 25]
$$


The maximum value is:

$$
Max = 25
$$


The minimum value is:

$$
Min = 10
$$


Therefore:

$$
Range = 25 - 10
$$


$$
Range = 15
$$


The distance between the smallest and largest values is **15**.

````python
import numpy as np

data = np.array([10, 12, 15, 18, 25])

maximum = np.max(data)
minimum = np.min(data)

data_range = maximum - minimum

print("Maximum:", maximum)
print("Minimum:", minimum)
print("Range:", data_range)
````

NumPy does not have a direct function called `range()` because Python already uses that name for generating sequences.

Therefore, we calculate the range by subtracting:

maximum value - minimum value

````python
data_range = np.ptp(data)

print(data_range)
````

`np.ptp()` means:

**Peak To Peak**

It calculates:

$$
Peak\ To\ Peak = Maximum - Minimum
$$

which is exactly the Range.

The range is very sensitive to extreme values.

Example:

Dataset without outlier:

$$
[10,12,15,18,25]
$$


Range:

$$
25-10=15
$$


After adding an outlier:

$$
[10,12,15,18,100]
$$


Range:

$$
100-10=90
$$


The range increased from **15 to 90** because of one extreme value.

````python
normal_data = np.array([10,12,15,18,25])

outlier_data = np.array([10,12,15,18,100])


print("Normal Range:", np.ptp(normal_data))

print("Range with Outlier:", np.ptp(outlier_data))
````

The range is useful for getting a quick idea about the spread of data.

However, because it only uses two values (minimum and maximum), it does not describe how the remaining values are distributed.

For real-world datasets, especially datasets containing outliers, measures like:

- Standard Deviation
- IQR

give a more reliable description of spread.

### Variance

Variance is a measure of **spread** that tells us how far the values in a dataset are distributed from the mean.

While the mean describes the **center** of the data, variance describes the **amount of variation** around that center.

A small variance means that data points are close to the mean.

A large variance means that data points are widely spread out.

Variance is calculated by taking the average of the **squared differences between each value and the mean**.

### Why Do We Need Variance?

Two datasets can have the same mean but completely different distributions.

Example:

Dataset A:

[10, 11, 12, 13, 14]

Dataset B:

[1, 5, 12, 19, 23]


Both datasets have a similar center, but Dataset B has much more spread.

Variance helps us measure this difference.

### Variance Formula

For a population:

$$
\sigma^2 = \frac{\sum (x_i-\mu)^2}{N}
$$


Where:

- $x_i$ = each value in the dataset
- $\mu$ = population mean
- $N$ = number of observations
- $\sigma^2$ = variance


The steps:

1. Calculate the mean.
2. Find the distance between each value and the mean.
3. Square each distance.
4. Find the average of squared distances.

### Example

Dataset:

[2, 4, 6, 8, 10]


Step 1: Calculate Mean

Mean = (2 + 4 + 6 + 8 + 10) / 5

Mean = 6


Step 2: Calculate deviations

2 - 6 = -4

4 - 6 = -2

6 - 6 = 0

8 - 6 = 2

10 - 6 = 4


Step 3: Square deviations

(-4)^2 = 16

(-2)^2 = 4

0^2 = 0

2^2 = 4

4^2 = 16


Step 4:

Variance = (16 + 4 + 0 + 4 + 16) / 5

Variance = 40 / 5

Variance = 8

````python
data = np.array([2,4,6,8,10])
variance = np.var(data)
print("Variance: " , variance)
````

### Checking the Steps with NumPy

We can calculate variance manually:

1. Find the mean.
2. Calculate squared differences.
3. Take the average.

````python
mean = np.mean(data)

squared_differences = (data - mean) ** 2

variance_manual = np.mean(squared_differences)

print("Mean:", mean)
print("Squared differences:", squared_differences)
print("Variance:", variance_manual)
````

### Variance and Outliers

Variance is highly affected by outliers.

Example:

Dataset:

[10, 12, 12, 13, 100]


The value 100 is far away from the other values.

Because variance uses squared differences, large distances become much larger.

Therefore:

- Outliers increase variance.
- Higher variance means more spread.
- Lower variance means more consistency.

````python
data_with_outlier = np.array([10, 12, 12, 13, 100])

print("Mean:", np.mean(data_with_outlier))
print("Variance:", np.var(data_with_outlier))
````

### Population Variance vs Sample Variance

In statistics, we distinguish between:

#### Population Variance

Used when we have the complete dataset.

Formula:

$$
\sigma^2 = \frac{\sum(x_i-\mu)^2}{N}
$$


#### Sample Variance

Used when we only have a sample from a larger population.

Formula:

$$
s^2 = \frac{\sum(x_i-\bar{x})^2}{n-1}
$$


The difference is the denominator:

Population:
- Divide by N


Sample:
- Divide by n-1


The n-1 correction prevents underestimating the true population variance.

````python
# Sample variance in NumPy

sample_variance = np.var(data, ddof=1)

print("Sample Variance:", sample_variance)
````

### Important Notes

- Variance is always non negative.
- Variance has squared units.

Example:

If the data is measured in meters:

Variance is measured in:

meters²


Because of squared units, variance is sometimes difficult to interpret directly.

This is why we use Standard Deviation next.

### Standard Deviation

Standard Deviation is a measure of **how spread out the data is around the mean**.

It tells us how far the values usually are from the average value.

Unlike variance, standard deviation is expressed in the **same units as the original data**, which makes it easier to interpret.

For example:

- If we measure height in centimeters, the standard deviation is also in centimeters.
- If we measure salary in dollars, the standard deviation is also in dollars.

A **small standard deviation** means that values are close to the mean.

A **large standard deviation** means that values are widely spread from the mean.


### Why do we need Standard Deviation?

Variance is useful mathematically, but it has one problem:

When we square the differences from the mean, the unit changes.

Example:

If data is measured in meters:

- Variance → meters²
- Standard Deviation → meters

Because of this, standard deviation is usually easier to understand and communicate.


### Formula

For a population:

$$
\sigma = \sqrt{\frac{\sum (x_i-\mu)^2}{N}}
$$


Where:

- $x_i$ = each value in the dataset
- $\mu$ = population mean
- $N$ = number of observations
- $\sigma$ = population standard deviation


The steps:

1. Calculate the mean.
2. Subtract the mean from each value.
3. Square each difference.
4. Calculate the average of squared differences.
5. Take the square root.


### Population vs Sample Standard Deviation

There are two types of standard deviation:

### 1. Population Standard Deviation

Used when we have the entire population.

Formula:

$$
\sigma = \sqrt{\frac{\sum(x_i-\mu)^2}{N}}
$$


### 2. Sample Standard Deviation

Used when we only have a sample from a larger population.

Formula:

$$
s = \sqrt{\frac{\sum(x_i-\bar{x})^2}{n-1}}
$$


The difference is the denominator:

- Population → $N$
- Sample → $n-1$

Using $n-1$ helps reduce bias when estimating the population spread from a sample.


### Example

Dataset:

$$
[10,12,12,13,100]
$$


##### Step 1: Calculate the mean

$$
Mean = \frac{10+12+12+13+100}{5}=29.4
$$


##### Step 2: Calculate deviations from the mean

| Value | Difference |
|---|---|
|10|10-29.4=-19.4|
|12|12-29.4=-17.4|
|12|12-29.4=-17.4|
|13|13-29.4=-16.4|
|100|100-29.4=70.6|


##### Step 3: Square the differences

Squared differences measure how far each value is from the mean.

##### Step 4: Average the squared differences

This gives the variance.

##### Step 5: Take the square root

This gives the standard deviation.


````python
import numpy as np

data = np.array([10, 12, 12, 13, 100])

# Mean
mean = np.mean(data)

# Variance
variance = np.var(data)

# Standard Deviation
std = np.std(data)

print("Mean:", mean)
print("Variance:", variance)
print("Standard Deviation:", std)
````

| Measure | Meaning | Unit |
|---|---|---|
| Variance | Average squared distance from the mean | Squared units |
| Standard Deviation | Typical distance from the mean | Same unit as data |


Example:

Dataset measured in kilograms:

Variance → kg²

Standard Deviation → kg

### IQR (Interquartile Range)

The **Interquartile Range (IQR)** is a measure of statistical spread that describes the range of the **middle 50% of the data**.

Unlike the range, IQR is **not affected strongly by outliers**, making it a robust measure of dispersion.

IQR measures the distance between the third quartile (Q3) and the first quartile (Q1).

$$
IQR = Q3 - Q1
$$

### Quartiles

Quartiles divide the dataset into four equal parts after sorting the values.

- **Q1 (First Quartile)**:
    - Also called the 25th percentile.
    - 25% of the data is below this value.

- **Q2 (Second Quartile)**:
    - The median.
    - 50% of the data is below this value.

- **Q3 (Third Quartile)**:
    - Also called the 75th percentile.
    - 75% of the data is below this value.


The relationship:

$$
Q1 \leq Median(Q2) \leq Q3
$$

The dataset is divided as:

| 25% | 25% | 25% | 25% |
|-----|-----|-----|-----|
| Min → Q1 | Q1 → Median | Median → Q3 | Q3 → Max |


The IQR represents only:

$$
Q1 \rightarrow Q3
$$

which contains the middle 50% of the observations.

Consider the following dataset:

$$
[10, 12, 15, 18, 20, 22, 25, 30, 35]
$$

The data is already sorted.

First quartile:

$$
Q1 = 15
$$

Third quartile:

$$
Q3 = 25
$$


Therefore:

$$
IQR = Q3 - Q1
$$


$$
IQR = 25 - 15 = 10
$$


The middle 50% of the data is spread across 10 units.

````python
import numpy as np

data = np.array([10, 12, 15, 18, 20, 22, 25, 30, 35])

# Calculate Q1 and Q3
Q1, Q3 = np.percentile(data, [25, 75])

# Calculate IQR
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
````

### Why IQR is Better Than Range?

Range:

$$
Range = Maximum - Minimum
$$


Example:

Dataset:

$$
[10,12,15,18,20,22,25,30,100]
$$


Range:

$$
100 - 10 = 90
$$


The value 100 (outlier) makes the range very large.


But IQR:

$$
Q3 - Q1
$$

ignores extreme values and focuses only on the middle 50% of the data.

| Measure | Formula | Sensitive to Outliers |
|---------|---------|----------------------|
| Range | Max - Min | Very sensitive |
| Variance | Average squared distance from mean | Sensitive |
| Standard Deviation | √Variance | Sensitive |
| IQR | Q3 - Q1 | Robust |


IQR is preferred when:
- The dataset contains outliers.
- The data is skewed.
- We want the typical spread of the majority of observations.

The IQR method is commonly used to detect outliers.

Lower boundary:

$$
Lower = Q1 - 1.5 \times IQR
$$


Upper boundary:

$$
Upper = Q3 + 1.5 \times IQR
$$


Any value:

$$
x < Lower
$$

or

$$
x > Upper
$$

is considered a potential outlier.

````python
import numpy as np

data = np.array([10, 12, 15, 18, 20, 22, 25, 30, 100])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[(data < lower_bound) | (data > upper_bound)]

print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
````

### Important Insight

An outlier is not always an error.

Before removing an outlier, we should investigate:

- Is it a data entry mistake?
- Is it a rare but real observation?
- Does it represent an important case?


The correct approach is:

**Detect → Investigate → Decide**

Never remove outliers automatically.

### Percentiles

A **percentile** is a value that indicates the percentage of data points that fall below it.

A percentile tells us the position of a value relative to the entire dataset.

- The 25th percentile means that 25% of the data is below this value.

Percentiles are useful because they help us understand the distribution and spread of data.

### Example

Consider the dataset:

[10, 20, 30, 40, 50]

The percentiles are:

- 25th percentile → value below which 25% of the data falls
- 50th percentile → median
- 75th percentile → value below which 75% of the data falls


````python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

q25 = np.percentile(data, 25)
q50 = np.percentile(data, 50)
q75 = np.percentile(data, 75)

print("25th percentile:", q25)
print("50th percentile:", q50)
print("75th percentile:", q75)
````

### Creating a Box Plot using Python

Box plots are useful because they allow us to quickly see:

- Center of data (median)
- Spread (IQR)
- Possible outliers
- Skewness of distribution


````python
import matplotlib.pyplot as plt

data = np.array([10, 12, 15, 18, 20, 22, 25, 30, 100])

plt.boxplot(data)

plt.title("Box Plot Example")
plt.ylabel("Values")

plt.show()
````

### Understanding Outliers using IQR Rule

An outlier is a value that is significantly different from the rest of the observations.

The IQR method identifies outliers using:

Lower Bound:

Q1 - 1.5 × IQR


Upper Bound:

Q3 + 1.5 × IQR


Any value:

- Smaller than the lower bound
- Larger than the upper bound

is considered a potential outlier.


### Detecting Outliers with Python

Steps:

1. Calculate Q1
2. Calculate Q3
3. Calculate IQR
4. Calculate lower and upper limits
5. Filter values outside these limits

````python
data = np.array([10, 12, 15, 18, 20, 22, 25, 30, 100])


Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1


lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR


outliers = data[(data < lower_bound) | (data > upper_bound)]


print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print("Outliers:", outliers)
````

### Important Analytical Decision

Outliers are not always errors.

An outlier can be:

1. A data entry mistake
2. A rare but real observation

Therefore:

We should investigate outliers before removing them.

Possible actions:

- Keep the value if it represents a real case
- Remove it if it is an error
- Transform the data
- Use robust measures like median and IQR
