## Bivariate Analysis

### What is Bivariate Analysis?

Bivariate Analysis is a statistical technique used to study the relationship between **two variables**.

Unlike **Univariate Analysis**, which focuses on a single variable, Bivariate Analysis examines how one variable changes with another.

The main objective is to answer questions such as:

- Is there a relationship between the two variables?
- How strong is the relationship?
- Is the relationship positive or negative?
- Can one variable help predict another?

For example, in the Medical Insurance dataset:

- Does age affect medical charges?
- Does BMI increase insurance cost?
- Do smokers pay more than non-smokers?

Instead of analyzing each column separately, we analyze the relationship between two columns.

---

### Types of Bivariate Analysis

The analysis method depends on the type of variables.

**1. Numerical vs Numerical**

Both variables are numerical.

Examples:

- Age vs Charges
- BMI vs Charges
- Height vs Weight
- Study Hours vs Exam Score

Common visualization:

- Scatter Plot

Common statistical method:

- Correlation Analysis

---

**2. Categorical vs Categorical**

Both variables are categorical.

Examples:

- Gender vs Smoking Status
- Gender vs Product Preference

Common methods:

- Contingency Table
- Chi-Square Test
- Count Plot

---

**3. Numerical vs Categorical**

One variable is numerical and the other is categorical.

Examples:

- Smoking Status vs Charges
- Gender vs BMI
- Education Level vs Income

Common visualization:

- Box Plot
- Violin Plot
- Bar Plot

---

### Why is Bivariate Analysis Important?

Bivariate Analysis helps us:

- Discover relationships between variables.
- Identify important features.
- Detect trends and patterns.
- Understand which variables influence the target variable.
- Prepare data before Machine Learning.

Without understanding variable relationships, building a predictive model becomes much more difficult.

---

### Univariate vs Bivariate vs Multivariate Analysis

| Analysis | Number of Variables | Purpose |
|-----------|---------------------|----------|
| Univariate | One | Understand a single variable |
| Bivariate | Two | Study the relationship between two variables |
| Multivariate | Three or more | Study interactions among multiple variables |

---

### Summary

Bivariate Analysis is the process of analyzing **two variables together** to understand their relationship.

Depending on the variable types, different visualization techniques and statistical methods are used.

It is one of the most important steps in Exploratory Data Analysis because it reveals hidden relationships inside the dataset.

## Scatter Plot


### What is a Scatter Plot?

A **Scatter Plot** is one of the most important visualization techniques in **Bivariate Analysis**.

It is used to visualize the relationship between **two numerical variables**.

Each point on the graph represents **one observation (one row)** in the dataset.

The x-axis represents one numerical variable.

The y-axis represents another numerical variable.

For example:

- Age vs Charges
- BMI vs Charges
- Height vs Weight

---

### Why do we use Scatter Plots?

Scatter plots help us:

- Discover relationships between variables.
- Detect positive or negative trends.
- Identify clusters.
- Detect outliers.
- Determine whether variables may be correlated.

It is usually the first visualization used before calculating a correlation coefficient.

---

### How Does a Scatter Plot Work?

Each observation in the dataset becomes one point.

Suppose we have the following data:

| Age | Charges |
|------|----------|
|19|16884|
|25|3200|
|32|4500|
|45|12000|

Each row is drawn as one point:

(19,16884)

(25,3200)

(32,4500)

(45,12000)

Hundreds or thousands of observations together form a pattern.

---

### Interpreting Scatter Plots

There are four common patterns.

**1. Positive Relationship**

As X increases,

Y also increases.

Example:

- Study Hours ↑
- Exam Score ↑

Or

- Age ↑
- Medical Charges ↑ (often)

The points move upward from left to right.

---

**2. Negative Relationship**

As X increases,

Y decreases.

Example:

- Product Price ↑
- Customer Demand ↓

The points move downward from left to right.

---

**3. No Relationship**

The points appear randomly scattered.

Knowing X tells us almost nothing about Y.

Example:

- Shoe Size
- IQ

---

**4. Non-linear Relationship**

Sometimes variables are related but **not in a straight line**.

Examples include:

- Quadratic curves
- Exponential growth
- Circular patterns

In these cases, the variables are related, but Pearson Correlation may not fully capture the relationship.

---

### Scatter Plot Example

Suppose we want to study:

Age vs Insurance Charges

If older people generally pay more insurance charges, the points will gradually move upward.

This suggests a positive relationship.

However, if the points are completely random, there is likely little or no relationship between age and charges.

---

### Advantages of Scatter Plot

- Very easy to understand.
- Reveals trends quickly.
- Shows outliers clearly.
- Helps detect correlation.
- Useful before building Machine Learning models.

---

### Limitations

- Works best with numerical variables.
- Large datasets may create overlapping points.
- Shows relationships visually but does not measure their strength.
- Cannot prove causation.

---

### Summary

A Scatter Plot is used to visualize the relationship between **two numerical variables**.

Each point represents one observation.

By examining the overall pattern of points, we can determine whether the relationship is:

- Positive
- Negative
- No relationship
- Non-linear

Scatter plots are usually followed by **Correlation Analysis**, which measures the strength of the relationship numerically.

---


### Scatter Plot Example

In this example, we study the relationship between **BMI** and **Insurance Charges**.

Additionally, we use the **Smoking Status** as a color grouping (`hue`) to compare smokers and non-smokers.

This allows us to examine whether smoking changes the relationship between BMI and insurance charges.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Data/insurance.csv")
df = df.drop_duplicates()

plt.figure(figsize=(7,5))
sns.scatterplot(data=df, x="bmi", y="charges", hue="smoker")
plt.xlabel("BMI")
plt.ylabel("Insurance Charges")
plt.title("BMI vs Charges, Colored by Smoking Status")
plt.show()
```

##### Explanation

- **x-axis (BMI):** Body Mass Index of each person.
- **y-axis (Charges):** Medical insurance charges.
- **Each point:** Represents one individual in the dataset.
- **Color (hue):** Indicates whether the individual is a smoker or a non-smoker.


##### Why use `hue`?

Using the `hue` parameter allows us to compare different categories within the same scatter plot.

Instead of displaying all observations with one color, Seaborn assigns a different color to each smoking category.

This makes it easier to identify patterns between smokers and non-smokers.

##### Observation

- Non-smokers are concentrated in a relatively narrow, lower range of insurance charges, regardless of their BMI values.
- Smokers are spread across a much wider range of insurance charges, with noticeably higher costs overall.
- As BMI increases, insurance charges tend to increase more significantly among smokers.
- This suggests that smoking has a strong influence on insurance charges, while the effect of BMI becomes more pronounced for smokers.


##### Interpretation

The scatter plot indicates that **Smoking Status** is a major factor affecting insurance charges.

Although BMI has some relationship with insurance charges, smokers consistently incur much higher medical costs than non-smokers with similar BMI values.

This suggests that smoking has a stronger impact on insurance charges than BMI alone and should be considered an important predictive feature.

## Correlation Analysis

---

### What is Correlation Analysis?

Correlation Analysis is a statistical technique used to measure the **strength** and **direction** of the relationship between two numerical variables.

While a Scatter Plot allows us to visually inspect the relationship, Correlation Analysis provides a numerical value called the **Correlation Coefficient (r)**.

The correlation coefficient ranges from **-1 to 1**.

---

### Pearson Correlation Coefficient

The most commonly used correlation measure is the **Pearson Correlation Coefficient**.

It measures the strength of a **linear relationship** between two numerical variables.

The value of **r** ranges between **-1** and **1**.

---

### Interpreting the Correlation Coefficient

| Correlation (r) | Interpretation |
|-----------------|----------------|
| 1.00 | Perfect Positive Correlation |
| 0.80 – 0.99 | Very Strong Positive |
| 0.50 – 0.79 | Strong Positive |
| 0.30 – 0.49 | Moderate Positive |
| 0.00 – 0.29 | Weak or No Correlation |
| 0 | No Linear Correlation |
| -0.30 – -0.49 | Moderate Negative |
| -0.50 – -0.79 | Strong Negative |
| -0.80 – -1.00 | Very Strong Negative |
| -1.00 | Perfect Negative Correlation |

---

### Types of Correlation

**Positive Correlation**

When one variable increases, the other variable also tends to increase.

Example:

- Study Hours ↑
- Exam Scores ↑

---

**Negative Correlation**

When one variable increases, the other tends to decrease.

Example:

- Product Price ↑
- Customer Demand ↓

---

**Zero Correlation**

There is no meaningful linear relationship between the variables.

Example:

- Shoe Size
- IQ


---

### Why is Correlation Analysis Important?

Correlation Analysis helps us:

- Measure relationships numerically.
- Identify important features.
- Detect redundant variables.
- Support feature selection before Machine Learning.
- Understand dependencies between variables.

---

### Correlation between BMI and Charges

```python


correlation = df["bmi"].corr(df["charges"])

print("Correlation:", correlation)
```

I can compute this for every pair of numeric columns at once with `.corr()`.

```python
corr = df.corr(numeric_only=True)
corr
```

**What this shows:** 

None of the numeric variables have a strong 
correlation with `charges` on their own - `age` (0.30) and `bmi` (0.20) 
are only weak-to-moderate. This actually makes sense given Steps 2-3: 
the real driver of high charges is `smoker`, which is categorical and 
doesn't show up in this numeric-only correlation table at all.

## Correlation Matrix & Heatmap

### What is a Correlation Matrix?

A **Correlation Matrix** is a table that shows the correlation coefficient between every pair of **numerical variables** in a dataset.

Instead of calculating the relationship between only two variables, it computes the correlation for **all numeric features at once**.

Each cell in the matrix contains a correlation coefficient ranging from **-1 to 1**.

---

### Why Do We Use a Correlation Matrix?

A correlation matrix helps us to:

- Understand relationships between numerical variables.
- Identify strongly correlated features.
- Detect redundant variables.
- Support feature selection before machine learning.
- Explore the dataset quickly.

---

### Understanding Correlation Values

The correlation coefficient always lies between **-1** and **1**.

| Correlation Value | Meaning |
|-------------------|---------|
| **+1** | Perfect positive correlation |
| **0** | No linear correlation |
| **-1** | Perfect negative correlation |

#### Interpretation

- **Positive correlation (+)** → Both variables increase (or decrease) together.
- **Negative correlation (-)** → When one variable increases, the other decreases.
- **Near zero (≈0)** → Little or no linear relationship.

> **Note:** The diagonal of the correlation matrix always contains **1.0**, because every variable is perfectly correlated with itself.

---

### Example

For the Medical Insurance dataset, the correlation matrix compares numerical variables such as:

- `age`
- `bmi`
- `children`
- `charges`

Each pair receives its own correlation coefficient.

## Heatmap

### What is a Heatmap?

A **Heatmap** is a color-coded visualization of the correlation matrix.

Instead of reading many numbers, colors make it much easier to identify strong and weak relationships.

---

### Reading a Heatmap

Typically:

- Dark warm colors → Strong positive correlation.
- Dark cool colors → Strong negative correlation.
- Neutral colors → Weak or no correlation.

The stronger the color, the stronger the relationship.

Heatmaps make large correlation matrices much easier to interpret.

```python
# Correlation Matrix

correlation_matrix = df.corr(numeric_only=True)

correlation_matrix
```

```python
# Heatmap

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Matrix Heatmap")
plt.show()
```

### Important Notes

- **`df.corr()`** calculates the correlation coefficients between all numerical columns.

- **`numeric_only=True`** ensures that only numerical columns are used and ignores text columns such as `sex` and `smoker`.

- **`annot=True`** displays the correlation value inside each square.

- **`cmap="coolwarm"`** selects a suitable color gradient to show positive and negative correlations.

- **`linewidths=0.5`** adds separating lines between cells to improve readability.

### General Interpretation Steps

1. Look for variables with high positive or negative correlation values.
   - These variables have stronger relationships.

2. Identify variables with very low correlation values.
   - They may have a weak linear relationship.

3. Consider the direction of the relationship:
   - Positive correlation means variables move in the same direction.
   - Negative correlation means variables move in opposite directions.

4. Use the heatmap to discover patterns and relationships before building machine learning models.

### Important Notes

- Correlation only measures **linear relationships** between variables.
- A low correlation value does not always mean that a variable is not important.
- Some important relationships may be non-linear and cannot be captured by Pearson correlation.
- Correlation does not imply causation; a relationship between two variables does not mean that one variable causes changes in the other.

### In Machine Learning

Correlation analysis can help with:
- Understanding relationships between features.
- Detecting highly correlated features (possible multicollinearity).
- Supporting feature selection decisions.

However, the heatmap should be used as an exploratory tool, not as the only reason to remove or keep features.

---

### Correlation Is Not Causation

Just because two variables move together doesn't mean one *causes* the 
other. There could be:
- A third variable driving both (a **confounder**)
- Pure coincidence
- The relationship actually running in the opposite direction

**Example from my own data:** `age` and `charges` have a correlation of 
0.30. That doesn't necessarily mean getting older *directly* causes 
higher charges - it could be that age is linked to other health factors 
I haven't measured here. EDA can show me *that* a relationship exists, 
but never *proves why* it exists. That's a question for a properly 
designed study or a controlled experiment, not a correlation coefficient.

## Pairplot

### What is Pairplot?

A **Pairplot** is a visualization technique provided by Seaborn that displays relationships between multiple numerical variables in a single grid.

Instead of creating separate scatter plots for each pair of variables, `sns.pairplot()` automatically creates all possible combinations.

---

### Why Do We Use Pairplot?

Pairplot helps us to:

- Quickly explore relationships between multiple variables.
- Identify possible correlations visually.
- Compare distributions of numerical features.
- Detect patterns, clusters, and outliers.
- Decide which relationships need deeper analysis.

---

### How Pairplot Works?

A pairplot contains:

- **Off-diagonal plots** → Scatter plots showing relationships between two variables.
- **Diagonal plots** → Distribution plots showing the shape of each variable.

Each row and column represents one numerical variable.


----

### Creating Pairplot

```python
sns.pairplot(df, vars=["age", "bmi", "charges"], hue="smoker")
plt.show()
```

#### Understanding the Output

The output is a **3×3 grid**:

Variables included:

- Age
- BMI
- Charges

Each row and column represents one variable.

The diagonal shows the distribution of each variable.

The other cells show the relationship between two variables using scatter plots.

The colors represent smoker status:

- Different colors separate smokers and non-smokers.
- This helps compare how smoking affects variable relationships.

#### Pairplot Insights

From the pairplot:

- The relationship between age and charges shows a general increasing trend.
- BMI and charges show a weaker relationship with more scattered points.
- Charges distribution is clearly different between smokers and non-smokers.
- Smokers tend to appear in regions with higher insurance charges.
- Pairplot provides a quick overview of relationships before performing deeper analysis.

#### Important Note 

`hue="smoker"` adds an extra dimension to the analysis:

Without `hue`:

We only see the relationship.

With `hue`:

We see the relationship broken down by category.

This is very important because `smoker` will often be one of the strongest variables explaining the difference in charges.

## Data Storytelling

### What is Data Storytelling?

Data Storytelling is the process of transforming data analysis results into a clear narrative that explains:

- What happened in the dataset.
- Why certain patterns appear.
- Which factors are important.
- What insights can be extracted from the analysis.

It combines:

- Data analysis.
- Visualizations.
- Domain understanding.

The goal is not only to show charts, but to explain the story behind the data.

---

### Why is Data Storytelling Important?

Data storytelling helps us to:

- Communicate findings clearly.
- Make data-driven decisions.
- Explain complex patterns to non-technical audiences.
- Connect EDA results with machine learning objectives.

A good analysis should answer:

1. What did we discover?
2. Why does it matter?
3. What action can be taken?
